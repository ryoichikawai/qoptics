# Dimer II: Strong coupling

In {numref}`ch:dimer-case1`, we considered the vanishing off-diagonal dissipators ($\Gamma_{12}=0$). In this chapter, we consider the opposite limit, $r_{12}/\lambda_{0} \rightarrow 0$, and thus  $\Gamma_{12} \rightarrow \gamma_{0}$.  Since the Dicke limit $r_{12}/\lambda_{0} = 0$ is unrealistic, we consider $r_{12} \sim 0.01\lambda_{0}$.  This is a case of triple time scales. (See {numref}`sec:openqm-qme-scaling`.)  The order of time scales are $\Omega/\omega_{0} \sim 10^{-5}$ and  $\gamma_{0}/\omega_{0} \sim 10^{-8}$.  This triple time scales requires the use of rotating frame with the Davies-GKSL master equation.

The density operator is computed in the interaction picture which captures the slow dissipative dynamics.  The exchange process between $|eg\rangle$ and $|ge\rangle$ won't show up in the interaction picture but we can recover it by rotating frame in backward. We will see that the effect of exchange process in the second-order coherence.
