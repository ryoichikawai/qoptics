# Dimer II: Coherence by indistinguishability

In {numref}`ch:dimer-case1`, we considered vanishing off-diagonal dissipators ($\Gamma_{12}=0$). In this chapter, we consider the opposite limit, $\Gamma_{12} \rightarrow \gamma_{0}$ where the distance between the two emitters is smaller than the wave length ($k_{0}r_{12} \ll 1$). The dipole-dipole coupling is not negligible and thus the degeneracy in the energy eigenvalues is lifted.

In this limit, interesting phenomenon such as superradiance, subradiance, and a dark state.
