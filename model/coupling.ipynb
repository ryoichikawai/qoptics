# Interaction between emitters and radiation field

For the emitters to emit and also to absorb a photon, they must interact with the radiation field. When the diameter of an emitter is much smaller than the wave length of radiation relevant to the optical processes under the investigation, we can take the electric-dipole approximation. Other forms of interaction such as the interaction between the angular momentum and the magnetic field are negligibly small.

## Electric-dipole (E1) approximation

Assuming that the wave length of light emitted by the atom is much larger than the size of the atom itself, electric-dipole approximation can be used and the interaction Hamiltonian can be written as

$$
H_{SR} = -\sum_{j}\mathbf{d}_j \cdot \mathbf{E}(\mathbf{x}_{j}) 
$$(eq:SR-coupling)

where  $\mathbf{d}_{j}$ is the dipole moment operator of the $j$-th atom and $\mathbf{x}_{j}$ is the center position of the atom. The dipole moment operator is defined by

$$
\mathbf{d}_{j} = \mathbf{d}  \sigma^{+}_{j} + \mathbf{d}^{*} \sigma^{-}_{j}
$$(eq:dipole-moment-op)

with  $\mathbf{d} = - \langle e| \mathbf{p} |g\rangle$ where $\mathbf{p}$ is the electric dipole moment of the atom, which is assumed to be identical to all the atoms. 

Substituting Eq. {eq}`eq:quantum-free-field-pm`, we have a full expression of the interaction Hamiltonian

$$
H_{SR} = - \sum_{j} \left[\mathbf{E}^{(+)}(\mathbf{x}_{j})+\mathbf{E}^{(-)}(\mathbf{x}_j)\right] \cdot \left( \mathbf{d}  \sigma^{+}_{j} + \mathbf{d}^{*} \sigma^{-}_{j} \right)
$$(eq:H_SR)

Recall that $E^{(-)}$ creates a photon and $E^{(+)}$ destroys one, and that $\sigma^{+}$ and $\sigma^{-}$ represents excitation and deexcitation, respectively.  Thenm $E^{(-)} \sigma^{-}$ and $E^{(+)} \sigma^{+}$ describe ordinary emission and absorption processes.  However, Eq. {eq}`eq:H_SR` contains two more terms, $E^{(-)} \sigma^{+}$ and $E^{(+)} \sigma^{-}$ which do not make a sense from the perspective of classical physics.  In quantum mechanics, emission and excitation can simultaneously happens despite the process does not conserve energy.  These processes take place during very short period of time and the Heisenberg uncertainty principle allows non-energy conserving process as long as the energy is conserved in a longer time scale. 

These virtual processes play key roles in various optical processes such as [spontaneous parametric down-conversion](https://en.wikipedia.org/wiki/Spontaneous_parametric_down-conversion) and [Raman scattering](https://en.wikipedia.org/wiki/Raman_scattering).  Their controibution to regular optical processes is small and we ignore them under the name of [rotating wave approximation](https://en.wikipedia.org/wiki/Rotating-wave_approximation).  The resulting coupling Hamiltonian is

$$
H_{SR}^\text{RWA} = - \sum_{j} \left[\mathbf{d} \cdot \mathbf{E}^{(+)}(\mathbf{x}_{j})\,   \sigma^{+}_{j} + \mathbf{d}^{*} \cdot \mathbf{E}^{(-)}(\mathbf{x}_j)\, \sigma^{-}_{j} \right]
$$(eq:H_SR-RWA)

## Dipole moments

The dipole moment is not a fixed property associated with an emitter.  It depends on many things, including the geometry around the emitter and the type of radiation field it interacts. It is considered as a vector.  That means it has a magnitude and direction.  If the emitter has a spherical symmetry and the radiation field is isotropic, how does a dipole moment pick a particular direction.  It can't.   There must be something else that breaks the symmetry.  We shall discuss it later and until then we assume that a coordinate system $x$, $y$, and $z$ are given.  As usual in quantum mechanics, we choose $z$ as the quantization axis and the energy eigen states carry angular momentum $(J, m)$.



### Definition of dipole moment operators

Suppose that there are $N$ charges $q_{i}, ], j=1,\cdots,N$ and there location is $\mathbf{r}_{j}$.  Then the dipole moment operator is defined as

$$
\hat{\mathbf{d}} = \sum_{j} q_{j} \hat{\mathbf{r}}_{j}
$$

where the caret "^" indicates quantum operator.  In our case, one electron ss responsible for light emission, $q=-e$ and

$$
\hat{\mathbf{d}} = -e \hat{\mathbf{r}}
$$

If the Cartesian coordinate is used, its components are 

$$
\hat{d}_{x} = -e \hat{x}, \quad \hat{d}_{y} = -e \hat{y}, \quad \hat{d}_{z} = -e \hat{z}
$$

Since we assume that the electronic state carries well defined angular momentum, it will be more convenient to express the dipole moment in the spherical coordinates

$$
\hat{d}_{0} = \hat{d}_{z}, \quad \hat{d}_{\pm 1} = \mp \frac{1}{\sqrt{2}} \left(\hat{d}_{x} + i \hat{d}_{z}\right)
$$

where the index $0, \pm 1$ comes from the spherical harmonics of degree 1:

$$
Y_{1}^{0} = -\sqrt{\frac{3}{4\pi}} \cos(\theta), \quad   Y_{1}^{\pm 1} = -\mp \sqrt{\frac{3}{8\pi}} \sin(\theta) e^{\pm i \phi} 
$$

Note that in the spherical coordinates, the dipole moment is complex.  

### Transition matrix elements and selection rules

For spherical emitter, we consider a transition from $|J,m\rangle$ to $|J',m'\rangle\rangle$.  The transition matrix is defined by

$$
\langle J',m'|\hat{\mathbf{d}}|J,m\rangle
$$

Using the spherical coordinates

$$
\langle J',m'|\hat{d}_{0}|J,m\rangle = d_{0}\,  \delta_{J,J'} \delta_{m,m'}, \qquad \langle J',m'|\hat{d}_{0}|J,m\rangle = d_{\pm 1}\,\delta_{m \pm 1,m'}
$$

There are two types of transitions, $\Delta m = 0$ and $\Delta m = \pm 1$.

### Liner ($\pi$) and circular ($\sigma^{\pm}$) transition dipoles

* $\Delta m = 0$:   The dipole moment is parallel to $z$ axis.  The emitted light has linear polarization in the $z$ direction.  Since the dipole moment is _parallel_ to the polarization, this transition is called $\pi$.
* $\Delta m = \pm 1$:  The dipole moment is in the $xy$ plane and it is a complex vector.  The emitted lights are polarized circularly one in a clockwise and the other in the opposite direction.  These transitions are called $\sigma_{\pm}$.

### Quantization axis

The above definition and characteristics of the dipole moment are all depend on the choices of the $z$ axis.  But who picks the $z$ axis?  That is entirely dependent on the experimental setup.  In practical applications, the emitters are largely in the same environment and thus it is natural to assume that they have the same dipole moment.

* dipole moment for a laser driving:   If an emitter is driven by a laser beam, its polarization determines the $z$ axis.  If the excitation does not change the value of $m$, then the laser should have a linear polarization, which couples to $d_{0}$.  Thus the $z$ direction is in the direction of the polarization. If $\Delta m = \pm 1$, then the laser must have have circular polarization.  The plane on which the polarization rotates define the $xy$ plane.  Thus the $z$ direction is in the direction of laser beam (the direction of $\mathbf{k}$).
* dipole moment for spontaneous emission and absorption:  If a spherical atom is isolated, then nothing fix the direction of dipole moment.  In that case, experimentalis often apply a weak magnetic field to break the spherical symmetry.  The direction of the magnetic field sets the $z$ direction. It is not necessary to fix $xy$ plane since the circular polarization rotates on it.  Defining a $z$ axis is enough for that.
* If the emitters are placed in a specific environment.  For example, quantum dots are either in a two-dimensional crystal or on a glass plate.  That sets a plane and a direction perpendicular to it.  The $z$ axis should be the one perpendicular to the plane.

### Non-spherical systems

Quantum dots are not exactly spherical but can be viewed as a sphere at least approximately.  So, the use of angular momentum can be justified.  However, the $z$ axis cannot be in an arbitral direction due to the underlying crystal structure.  It is not quire sure exactly what direction should be used.  It is also not certain that an ensemble of emitters have the same direction of dipole moments.
