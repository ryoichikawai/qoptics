Using the free Hamiltonian {eq}`eq:H0-dimer`, the dipole-dipole coupling {eq}`eq:H_dd-2` and the disspator {eq}`eq:dimer-case1-dissipator`, we solve the quantum master equation using `QuTiP`.

## Steady state


```python
# Parameters
omega0 = 1.0
gamma0 = 1e-8
T = 0.86*omega0 
k0, lambda0 = scale(omega0)
a = 2*lambda0  # distance between two emitters

r12 = a*2*np.pi # k0*r12
theta_dipole = np.pi/2

Omega = ddcoupling(r12,theta_dipole)
print("** Magnitude of dipole-dipole coupling **")
print("   Omega/omega0 = {0:10.4e},   Omega/gamma0 = {1:7.4f}".format(Omega*gamma0/omega0,Omega))
H = hamiltonian(Omega)
nbar = planck_dist(omega0,T)
gamma = decay_rate(r12,theta_dipole)
L, c_ops = collapse_ops(gamma,nbar)

# Steady state
rho_ss = steadystate(H,c_ops)
```

First we check if the steady state follows the Boltzmann distribution.  The numerical results agree with the theory perfectly. Then, we also check if no coherence between $|eg\rangle$ and $|ge\rangle$.   The corresponding matrix element indeed vanished. 

```python
# Gibbs state
W, E, order = eigen_energies(omega0,Omega,gamma0)
P = get_Gibbs(E,T)

print("** Diagonal elements of density operator **")
print("          numerical theoretical")
print("ρ_ee_ee  = {0:7.4f}  {1:8.4f}".format(rho_ss[0,0].real,P[0]))
print("ρ_eg_eg  = {0:7.4f}  {1:8.4f}".format(rho_ss[1,1].real,P[1]))
print("ρ_ge_ge  = {0:7.4f}  {1:8.4f}".format(rho_ss[2,2].real,P[2]))
print("ρ_gg_gg  = {0:7.4f}  {1:8.4f}".format(rho_ss[3,3].real,P[3]))

print("\n** Off-diagonal elements of density operator **")
print("ρ_eg_ge  = {0:7.4f}".format(rho_ss[2,1]))
```

As a further check if cross correlation indeed vanishes as we assumed.  The following code confirms it.

```python
sz, sp, sm = spin_ops()
x=expect(sp[0]*sm[0],rho_ss)
print("<sigmap_1 sigmam_1> = {0:7.4f}".format(x))
x=expect(sp[1]*sm[1],rho_ss)
print("<sigmap_2 sigmam_2> = {0:7.4f}".format(x))
x=expect(sp[0]*sm[1],rho_ss)
print("<sigmap_1 sigmam_2> = {0:7.4f}".format(x))
```

## First-order coherence function (Detector Model I)

If detector model I is used, $g^{(1)}(\tau)$ should be identical to the single emitter case.

```python
times = np.linspace(0, 5, 500)
theta_detect = np.pi/2
phi_detect = 0
nhat = detector_pos(theta_detect,phi_detect)
r=emitter_pos(r12)

D1=np.exp(-1j*np.inner(nhat,r[0])) * sm[0]
D2=np.exp(-1j*np.inner(nhat,r[1])) * sm[1]
D = [D1,D2]

I1 = 0
G1 = 0
for k in range(2):
    rho_0 = rho_ss*D[k].dag()
    e_ops = D[k]
    result = mesolve(H,rho_0,times,c_ops,e_ops=e_ops)
    I1 += expect(D[k].dag()*D[k],rho_ss)
    G1 += result.expect[0]

g1 = G1/I1
g1_th = np.exp(-times)
plt.figure(figsize=(4,3))
plt.title("Detector Model I")
plt.plot(times,g1.real,color="b",label="mesolve")
plt.plot(-times,g1.real,color="b")
plt.plot(times,g1_th.real,color="y",ls='--',label="theory")
plt.plot(-times,g1_th.real,color="y",ls='--')
plt.xlabel(r"$\tau\gamma_{0}$")
plt.ylabel(r"$g^{(1)}(\tau)$")
plt.legend(loc=1)
plt.show()
```

## First-order coherence function (Detector Model II)

The  first-order coherence function should not depend on the detector model.  The two results are identical.

```python
DD = sum(D)

# intensity
I2=expect(DD.dag()*DD,rho_ss)

rho_0 = rho_ss*DD.dag()
e_ops = DD
result = mesolve(H,rho_0,times,c_ops,e_ops=e_ops)
G1 = result.expect[0]

g1 = G1/I2
g1_th = np.exp(-times)
plt.figure(figsize=(4,3))
plt.title("Detector Model II")
plt.plot(times,g1.real,color="b",label="mesolve")
plt.plot(-times,g1.real,color="b")
plt.plot(times,g1_th.real,color="y",ls='--',label="theory")
plt.plot(-times,g1_th.real,color="y",ls='--')
plt.xlabel(r"$\tau\gamma_{0}$")
plt.ylabel(r"$g^{(1)}(\tau)$")
plt.legend(loc=1)
plt.show()
```

## Second-order coherence function (Detector model I)

The coherence function measured by a detector does not depend on the location as shown in xxx.  It is the same as emission counting without interference.  Nevertheless, $g^{(2)}(\tau)$ from the incoherent pair is not the same as that of a single emitter.  The no-delay coherence is $g^{(2)}(0)=0.5$ instead of $0$ as predicted by Eq. {eq}`eq:g2-incoherent-N`, suggesting that there is a good chance that two photons arrive at the same time.  When a photon emitted by the first emitter, the same emitter can't emit the second photon immediately.  However, another photon from the second emitter can arrive at the same time as the first one since the two photons emitted by different emitters are completely independent from each other.  The following code clearly shows it.  In next section, we unravel the process using the jump method.

```python
G2 = 0
for i in range(2):
    for j in range(2):
        
        rho_0 = D[i]*rho_ss*D[i].dag()
        e_ops = D[j].dag()*D[j]
        result = mesolve(H,rho_0,times,c_ops,e_ops=e_ops)
        G2 += result.expect[0]      

g2 = G2/I1**2
g2_th = 1 - 0.5*np.exp(-(2*nbar+1)*times)
plt.figure(figsize=(4,3))
plt.title("Detector Model I")
plt.plot(times, np.real(g2),c='r',label='mesolve')
plt.plot(-times,np.real(g2),c='r')
plt.plot(times,np.real(g2_th),c='y',ls='--',label='theory')
plt.plot(-times,np.real(g2_th),c='y',ls='--')
plt.xlabel(r'$\tau\gamma_{0}$')
plt.ylabel(r'$g^{(2)}(\tau)$')
plt.axhline(y=0.5,color='grey',ls='--')
plt.ylim([0,1.2])
plt.legend(loc=4)
plt.grid(True)
plt.show()
```

## Second-order coherence function (Detector model II)

```python

rho_0 = DD*rho_ss*DD.dag()
e_ops = DD.dag()*DD
result = mesolve(H,rho_0,times,c_ops,e_ops=e_ops)
G2 = result.expect[0]      

g2 = G2/I2**2

plt.figure(figsize=(4,3))
plt.title("Detector Model II")
plt.plot(times, np.real(g2),c='r')
plt.plot(-times,np.real(g2),c='r')
plt.xlabel(r'$\tau\gamma_{0}$')
plt.ylabel(r'$g^{(2)}(\tau)$')
plt.ylim([0,1.5])
plt.grid(True)
plt.show()
```
