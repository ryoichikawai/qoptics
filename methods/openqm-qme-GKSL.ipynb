# Davies-GKSL master equation

## Issues with the optical quantum master equation

The quantum master equation is based on the second order perturbation theory along with Markovian and other approximations.  It is not always perfect.  When the interaction with environment is strong, it obviously fails.  Fortunately, interaction between emitters and radiation fields is generally weak.  It works reasonably well. However, there are situations where the results are qualitatively unacceptable.  We already mentioned that the dipole-dipole coupling diverges under the Decke limit.  In that case, we stays in the safe parameter regions, that is $r_{12} > 0.01 \lambda_{0}$.  Even in this parameter region,  the standard optical master equation could fail.  Here we discuss one of such issues.

The definition of $\Omega_{ij}$ and $\Gamma_{ij}$ in Eqs. {eq}`eq:omega_ij` and {eq}`eq:Gamma_ij`, we used the free decay rate $\gamma_0$ using a single Borh frequency $\omega_0$.  When emtters are free (no mutual intreaction nor external potential), that is correct.  However, when the free Hamiltonian is perturbed by the coupling, the transition energy is no longer $\omega_{0}$.  There are multiple Bohr frequencies and each transition should have different transition rates. But the master equation {eq}`eq:optical-qme` uses a single Bohr frequency $\omega_{0}$ in any case regardless of the actual Bohr frequancies.`

The question is whether such approximation is acceptable or not.  If the omission of the change in the Bohr frequencies causes only small quantitative errors, we can accept it.  However, if the error cause qualitative difference, then we need to use a better approximation.  Indeed the quantitative change of $\Omega_{12}$ and $\Gamma_{12}$ is not very big.  However, it causes qualitative error from which leads us to a wrong conclusion.

The key issue is the broken detailed balance.  If only one Bohr frequency is used, the steady state is always diagonal in the product basis and the system never reaches the true thermal state (Gibbs state).  For the two emitters, coherence between $|eg\rangle$ and $|ge\rangle$ will never be established by the dipole-dipole interaction if {eq}`eq:omega_ij`
and {eq}`eq:Gamma_ij`.

If we want to investigate the coherence between the product basis by the interaction between the emitters, we need to use a different method.  


## GKSL master equation

It is possible to construct a quantum master equation that satisfies the detailed balance.  The basic theory was developed in {cite:p}`Davies1974,Gorini1976,Lindblad1976`, which is known as the GKSL master equation.  Microscopic derivation of the GKSL optical master equation is given in {cite:p}`Breuer2002`.

Consider the Bohr frequency $\omega_{\alpha\beta}=E_{\alpha}-E_{\beta}$ between energy eigenstate $|E_{\alpha}\rangle$ and $|E_{\beta}\rangle$.  Then, we define jump operators for emission

$$
A_{j}(\omega_{\alpha\beta}) = |E_{\beta}\rangle\langle E_{\beta}| \sigma^{-}_{j}|E_{\alpha}\rangle\langle E_{\alpha}|
$$

and for absorption

$$
A_{j}(\omega_{\beta\alpha}) =  |E_{\alpha}\rangle\langle E_{\alpha}| \sigma^{+}_{j}|E_{\beta}\rangle\langle E_{\beta}|.
$$

Noting that $\omega_{\alpha\beta} = - \omega_{\beta\alpha}$, we have $A^{\dagger}(\omega_{\alpha\beta})=A(\omega_{\beta\alpha}) = A(-\omega_{\alpha\beta})$.

and the dissipator is defined as

$$
\begin{aligned}
D(\rho) =& \sum_{ij} \sum_{\omega_{\alpha\beta}>0} \Gamma_{ij}^{\downarrow}(\omega_{\alpha\beta}) \left[ A_{i}(\omega_{\alpha\beta}) \rho A^{\dagger}_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A^{\dagger}_{j}(\omega_{\alpha\beta})  A_{i}(\omega_{\alpha\beta}),\rho\right\}\right] \\
& + \sum_{ij} \sum_{\omega_{\alpha\beta}>0} \Gamma_{ij}^{\uparrow}(\omega_{\alpha\beta}) \left[ A^{\dagger}_{i}(\omega_{\alpha\beta}) \rho A_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A_{j}(\omega_{\alpha\beta})  A^{\dagger}_{i}(\omega_{\alpha\beta}),\rho\right\}\right] 
\end{aligned}
$$(eq:GKSL-dissipator)

where emission rate $\Gamma^{\downarrow}_{ij}$ and absorption rate $\Gamma^{\uparrow}_{ij}$ are defined by 

$$
\begin{aligned}
\Gamma^{\downarrow}_{ij} (\omega_{\alpha\beta}) &= \frac{4 d^2 \omega_{\alpha\beta}^3}{3 \hbar c} \left[1 + N(\omega_{\alpha\beta})\right]\\
\Gamma^{\uparrow}_{ij} (\omega_{\alpha\beta}) &= \frac{4 d^2 \omega_{\alpha\beta}^3}{3 \hbar c} N(\omega_{\alpha\beta})
\end{aligned}
$$(eq:GKSL-Gamma)

with $N(\omega) = \left(e^{\hbar \omega/T}-1\right)^{-1}$.

These transition rate satisfies the detailed balance 

$$
\frac{\Gamma^{\uparrow}(\omega_{\alpha\beta})}{\Gamma^{\downarrow}(\omega_{\alpha\beta})} = \frac{N(\omega_{\alpha\beta})}{1 + N(\omega_{\alpha\beta})} = e^{-\omega_{\alpha\beta}/T}
$$

which guarantees that the steady state is the Gibbs state.

In conclusion, when the dipole-dipole coupling is weak, that is $\Omega_{ij} \ll \omega_{0}$ and also $\Omega_{ij} \ll T$, we use the standard optical master equation. Otherwise, we use the GKSL version with {eq}`eq:GKSL-dissipator` and {eq}`eq:GKSL-Gamma`.

## The number of jump operators

In general, there is a jump operator for each transition.  That means For N=2, there for 4 jump operators for emission and another 4 operators for absorption, totaling 8 operators.  The standard optical master equation had only 4.  For N=3 with $S_3$ symmetry,  we need 14 operators and if there is no symmetry there are 26 operators. The optical master equation needs only 6.   The number of jump operators grows very rapidly as the number of emitters increases.  Hence, this method is not practical for large $N$ unless the system has a high level of symmetry.
