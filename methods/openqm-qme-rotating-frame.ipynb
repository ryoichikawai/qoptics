# Rotating frame

As discussed in {numref}`sec:openqm-qme-scaling`, there are three time scales set by $\omega_{0}$, $\Omega$ and $\gamma_{0}$.  Depending on the strength of the dipole-dipole coupling, we have two cases, dual and triple time scales.   For the dual time scale, we just need to eliminate $\omega_{0}$ from the master equation since $\Omega$ is negligibly small.  Then we can move to a rotating frame using a unitary transformation $U_\text{I} = e^{-i H_{0} t}$ (Interaction Picture I).   If the interaction is strong compared with $\gamma_{0}$, we need to eliminate both $\omega_{0}$ and $\Omega$ using $U_\text{II} = e^{- i (H_{0}+V_\text{dd})t}$ (Interaction Picture II), which requires the use of Davies-GKSL mastre equation.

## Interaction picture I

We consider the standard optical master equation

$$
\frac{d}{dt} \rho = -i \left[H_{0}+V_{dd},\rho\right] + \mathcal{D}^{\downarrow}(\rho) + \mathcal{D}^{\uparrow}(\rho)
$$

where

$$
\begin{aligned}
H_{0} &= \frac{\omega_{0}}{2} \sum_{j} \sigma^{Z}_{j} \\
V_{dd} &= \sum_{i \ne j} \Omega_{ij} \sigma^{+}_{i} \sigma^{-}_{j} \\
\mathcal{D}^{\downarrow}(\rho) &= \sum_{ij} \Gamma_{ij}^{\downarrow} \left( \sigma^{-}_{j} \rho \sigma^{+}_{i} - \frac{1}{2}\left\{\sigma^{+}_{i} \sigma^{-}_{j}\right\}\right) \\
\mathcal{D}^{\uparrow}(\rho) &= \sum_{ij} \Gamma_{ij}^{\uparrow} \left( \sigma^{+}_{j} \rho \sigma^{-}_{i} - \frac{1}{2}\left\{\sigma^{-}_{i} \sigma^{+}_{j}\right\} \right) \\
\end{aligned}
$$

We want to eliminate $\omega_{0}$ from the equation by using a unitary transformation $U_\text{I} = e^{-i H_{0} t}$.  Introducing the interaction picture

$$
\tilde{\rho} = U_\text{I}^{\dagger} \rho U_\text{I}
$$(eq:rho-int-pict1)

and 

$$
\tilde{\sigma}^{\pm}_{j} = U_\text{I}^{\dagger} \sigma^{\pm}_{j} U_\text{I} = e^{\pm i \omega_{0} t} \sigma^{\pm}_j
$$(eq:sigma-int-pict1)

The relation {eq}`eq:sigma-int-pict1` is crucial since the dependence on $\omega_{0}$ are canceled if ^\sigma^{+}$ and $\sigma^{-}$ appear as a pair.  Othre wise, $\omega_{0}$ just moved from the Hamiltonian to the dissipator.

The master equation becomes

$$
\frac{d}{dt} \tilde{\rho} = -i \left[V_{dd},\tilde{\rho}\right] + \mathcal{D}^{\downarrow}(\tilde{\rho}) + \mathcal{D}^{\uparrow}(\tilde{\rho})
$$(eq:qme-int-pict1)

Notice that the free Hamiltonian $H_{0}$ vanished. Since $\tilde{\sigma}^{\pm}_{j} \tilde{\sigma}^{\mp}_{j} = \sigma^{\pm}_{j} \sigma^{\mp}_{j}$, neither $V_{dd}$ nor $\mathcal{D}(\cdot)$ changes by the transformation.  Furthermore, expectation values are invariant under the unitary transformation and thus we do have to change the coherence functions as well.  This is mathematically exact and not an approximation.    Since $\omega_{0}$ is no longer present, we have one time scale set by $\gamma_{0}$ which appears both in $\Omega_{ij}$ and $\Gamma_{ij}$ (See {numref}`sec:openqm-qme-optical`.)   Now scaling time as $\tau = \gamma_{0} t$, the master equation {eq}`eq:qme-int-pict1` becomes


$$
\frac{d}{d\tau} \tilde{\rho} = -i \left[V_{dd}/\gamma_{0},\tilde{\rho}\right] + \mathcal{D}^{\downarrow}(\tilde{\rho})/\gamma_{0} + \mathcal{D}^{\uparrow}(\tilde{\rho})/\gamma_{0}
$$(eq:qme-int-pict1-tau)

Now, no small or large number is involved in the equation. 

## Interaction picture II

For the case of triple time scales, we eliminate the entire Hamiltonian using $U_\text{II}$.  Similarly to Eq. {eq}`eq:rho-int-pict1`, the density operator in the interaction picture is given by

$$
\tilde{\rho} = U_\text{II}^{\dagger} \rho U_\text{II}
$$

However, the relation {eq}`eq:sigma-int-pict1` is no longer work under this interaction picture.  We recall that the jump operators in the Davies-GKSL master equation satisfy

$$
U_\text{II}^{\dagger}  A_{j}(\omega) U_\text{II} = e^{-i \omega t} A_{j}(\omega)
$$

which is equivalent to  Eq. {eq}`eq:sigma-int-pict1`.  AS long as $A(\omega)$ and $A^{\dagger}(\omega)$ appear as a pair, the fast oscillating factors cancel out.  Thyen, the equation of motion is given by

$$
\frac{d}{d\tau} \tilde{\rho} =  \mathcal{D}^{\downarrow}(\tilde{\rho}) + \mathcal{D}^{\uparrow}(\tilde{\rho})
$$(eq:qme-int-pict2)

where

$$
\begin{aligned}
\mathcal{D}^{\downarrow} (\tilde{\rho}) &=\sum_{ij} \sum_{\omega_{\alpha\beta}>0} \frac{\Gamma_{ij}^{\downarrow}(\omega_{\alpha\beta})}{\gamma_{0}} \left[ A_{i}(\omega_{\alpha\beta}) \tilde{\rho} A^{\dagger}_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A^{\dagger}_{j}(\omega_{\alpha\beta})  A_{i}(\omega_{\alpha\beta}),\tilde{\rho}\right\}\right] \\
\mathcal{D}^{\uparrow} (\tilde{\rho}) &= \sum_{ij} \sum_{\omega_{\alpha\beta}>0} \frac{\Gamma_{ij}^{\uparrow}(\omega_{\alpha\beta})}{\gamma_{0}} \left[ A^{\dagger}_{i}(\omega_{\alpha\beta}) \tilde{\rho} A_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A_{j}(\omega_{\alpha\beta})  A^{\dagger}_{i}(\omega_{\alpha\beta}),\tilde{\rho}\right\}\right] 
\end{aligned}
$$

There is no need to use different operators in the coherence functions.  Solve the master equation {eq}`eq:qme-int-pict2` with the dimensionless time $t\gamma_{0}$.
