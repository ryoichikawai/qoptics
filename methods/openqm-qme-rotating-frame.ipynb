# Rotating frame

As discussed in {numref}`sec:openqm-parameters`, there are two time scales determined by $\omega_{0}$ and $\gamma_{0}$.  Their ratio is as small as $\gamma_{0}/\omega_{0} \sim 10^{-8}$.  It is not possible to take into account the two scale in numerical calculation since the system oscillates 100 million times during decay time.  Fortunately, we can separate the time scales easily using the rotating frame. 

## Interaction picture

We consider the standard optical master equation

$$
\frac{d}{dt} \rho = -i \left[H_{0}+V_{dd},\rho\right] + \mathcal{D}^{\downarrow}(\rho) + \mathcal{D}^{\uparrow}(\rho)
$$

where

$$
\begin{aligned}
H_{0} &= \frac{\omega_{0}}{2} \sum_{j} \sigma^{Z}_{j} \\
V_{dd} &= \sum_{i \ne j} \Omega_{ij} \sigma^{+}_{i} \sigma^{-}_{j} \\
\mathcal{D}^{\downarrow}(\rho) &= \sum_{ij} \Gamma_{ij}^{\downarrow} \left( \sigma^{-}_{j} \rho \sigma^{+}_{i} - \frac{1}{2}\left\{\sigma^{+}_{i} \sigma^{-}_{j}\right\}\right) \\
\mathcal{D}^{\uparrow}(\rho) &= \sum_{ij} \Gamma_{ij}^{\uparrow} \left( \sigma^{+}_{j} \rho \sigma^{-}_{i} - \frac{1}{2}\left\{\sigma^{-}_{i} \sigma^{+}_{j}\right\} \right) \\
\end{aligned}
$$

We want to eliminate $\omega_{0}$ from the equation by using a unitary transformation $U = e^{-i H_{0} t}$.  Introducing the interaction picture

$$
\tilde{\rho} = U^{\dagger} \rho U
$$

and 

$$
\tilde{\sigma}^{\pm}_{j} = U^{\dagger} \sigma^{\pm}_{j} U = e^{\pm i \omega_{0} t} \sigma^{\pm}_j
$$

The master equation becomes

$$
\frac{d}{dt} \tilde{\rho} = -i \left[V_{dd},\tilde{\rho}\right] + \mathcal{D}^{\downarrow}(\tilde{\rho}) + \mathcal{D}^{\uparrow}(\tilde{\rho})
$$(eq:qme-rotating-frame)

Notice that the free Hamiltonian $H_{0}$ vanished. Since $\tilde{\sigma}^{\pm}_{j} \tilde{\sigma}^{\mp}_{j} = \sigma^{\pm}_{j} \sigma^{\mp}_{j}$, neither $V_{dd}$ nor $\mathcal{D}(\cdot)$ changes by the transformation.  Furthermore, expectation values are invariant under the unitary transformation and thus we do have to change the coherence functions as well.  This is mathematically exact and not an approximation.    Since $\omega_{0}$ is no longer present, we have one time scale set by $\gamma_{0}$ which appears both in $\Omega_{ij}$ and $\Gamma_{ij}$ (See {numref}`sec:openqm-qme-optical`.)   Now scaling time as $\tau = \gamma_{0} t$, the master equation {eq}`eq:qme-rotating-frame` becomes


$$
\frac{d}{d\tau} \tilde{\rho} = -i \left[V_{dd}/\gamma_{0},\tilde{\rho}\right] + \mathcal{D}^{\downarrow}(\tilde{\rho})/\gamma_{0} + \mathcal{D}^{\uparrow}(\tilde{\rho})/\gamma_{0}
$$(eq:qme-rotating-frame2)

Now, no small or large number is involved in the equation.  Hence, it is numerically tractable.

## Role of the dipole-dipole coupling

Since the dipole-dipole coupling is very small compared to $\omega_{0}$,  the energy shift due to the coupling can be completely neglected.  However, in the rotating frame, it is not small after time is scaled by  $\gamma_{0}$.  That means the dipole-dipole coupling contributes to the relaxation process.  The coherence functions measure the relaxation.  Thus, we need to include the dipole-dipole coupling in the calculation of coherence functions.
