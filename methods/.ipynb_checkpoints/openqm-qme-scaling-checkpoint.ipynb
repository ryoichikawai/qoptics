# Parameter values and normalization

## Essential parameters

There are five parameters for every cases we study.

1. wave length:  $\lambda_0$.
2. magnitude of dipole moment: $d$ $\quad\Rightarrow\quad$ the free spontaneous emission rate: $\gamma_{0}$.
3. temperature of a photon gas: $T$.
4. angle between the dipole moment and the line connecting two emitters:  $\chi$.
5. unit vector from the center of emitters to the detector: $\hat{\mathbf{n}}$ specifyed by polar angles: $\theta$ and $\phi$.
6. location of each emitter: $\mathbf{r}_{i}$.
7. strength of external driving $E_{L}$  $\quad\Rightarrow\quad$  Rabi frequency $\Omega_{L} = \mathbf{d}\cdot\mathbf{E}_{L}$  ({numref}`sec:driving`)

_Remarks_
1. Instead of specifying the magnitude of dipole moment, we use the free spontaneous emission rate $\gamma_{0} = \displaystyle\frac{4 \omega_{0}^{3} d^{2}}{3 \hbar c^2}$ as basic parameter.
2. Similarly, the Rabi frequency $\Omega_{L} = \mathbf{d}\cdot\mathbf{E}_{L}$ is used as a parameter instead of the laser intensity.


```{important} Basic parameters

1. Bohr frequency: $\omega_{0}$
2. Spontaneous emission rate: $\gamma_{0}$
3. temperature:  $T$
4. dipolemoment direction:  $\chi$
5. detector location:  $(\theta, \phi)$
6. location of emitters:  $\mathbf{r}_{i},\,  i=1, \cdots, N$
7. Rabi frequency $\Omega_{L}$.

```

## Length scale and unit 

We have only one length scale, which is the distance between emitters.  It is convenient to measure the distance with respect to the wave length of light.  For a given value of $\omega_{0}$, the wave number and wave length  are

$$
\lambda_{0} = \frac{2\pi}{k_{0}} = \frac{2\pi c}{\omega_{0}}
$$

In this note, we assume $c=1$.  Hence, $\omega_{0} = k_{0} = \displaystyle \frac{2\pi}{\lambda_{0}}$.

Let $r^\text{L}$ is the distance in the laboratory unit. In numerical calculation, we use normalized distance $r = k_{0} r^{\text{L}}$.  In plotting, we use more intuitive unit $r^\prime = r^\text{L}/\lambda_{0}$.    Both $r$ and $r^\prime$ are dimensionless and related by $r = 2 \pi r^\prime$.

## Time scale and unit

There are two time scales, $\omega_{0}$ and $\gamma_{0}$. The frequency $\omega_{0}$ dictates the rapid oscillation and the decay rate $\gamma_{0}$ represents a very slow relaxation process.  In a typical case, $\displaystyle \frac{\gamma_{0}}{\omega_{0}} \sim 10^{-8}$.   Since the their magnitiude of order are far apart, we treat the two time scales separately.  Particularly when we calculate time evolution of the system state, the duration of the process must be longer than $\gamma_{0}^{-1}$.  Then, we need to remove the rapid oscillation from the dynamics.  It can be done by using a rotating frame.  See xxxx.

In numerical calculation, we use two times.  When we are interested in sort time behavior with in several phase oscillations, $t = \omega_{0}^{L} t^\text{L}$  where $\omega^\text{L}_{0}$ and $t^\text{L}$ is the frequency and time in laboratory unit.  On the other hand, for a slow relaxation dynamics, we use  $t^\prime = \gamma_{0}^\text{L} t^\text{L} = \displaystyle\frac{\gamma_{0}^\text{L}}{\omega_{0}^{L}} t = \gamma_{0} t$ where we defined $\gamma_{0} =   \displaystyle\frac{\gamma_{0}^\text{L}}{\omega_{0}^{L}}$.

When external driving is used, the Rabi frequency $\Omega_{L}$ sets another time scale.

## Energy scale and unit

There are two energy scales,, one is the transition energy $\hbar \omega_{0}$ or equivalently the energy of photon.  We assume $\hbar = 1$.  Temperature should be similar to the transition energy.  The second energy scale is the dipole-dipole coupling strength $\Omega$ which is in the similar order to $\gamma_{0}$.  Since $\displaystyle\frac{\Omega}{\omega_{0}}$ is so small, the dipole-dipole coupling can be ignored from the energy point of view.  However, during a slow relaxation process, the coupling can mae a significant contribution.  Hence, $\displaystyle\frac{\Omega}{\gamma_{0}} \sim \mathcal{O}(1)$ is more relevant. 

Like the time scale, we use two energy scales separately.  For energy transaction we express energy as $E = \displaystyle\frac{E^\text{L}}{\hbar\omega_{0}^\text{L}}$, which means we set $\omega_{0}=1$ in numerical calculation.  Other energy such as temperature is measured with respect to $\omega_{0}$.  For time integration, the energy (Hamiltonian) is scaled with $\gamma_{0}$ like $\Omega = \displaystyle\frac{\Omega^\text{L}}{\gamma_{0}^\text{L}}$.

```{important} Dimensionless quantities

1. length:  $k_{0} r$  or $r/\lambda_{0}$
2. time: $\omega_{0} t$ or $\gamma_{0} t$
3. energy:  $E/\omega_{0}$

```

## Typical values

There is no typical value for angles, $\chi$, \theta$, and $\phi$.

We assume the following realistic values

1. $\lambda_{0} \sim 600 $ nm  $\quad \Rightarrow\quad \omega_{0} \sim 3 \times 10^{15} \text{s}^{-1} \Rightarrow 0.3 \times 10^{-15} \test {s} = 0.3 \text{ ps}.
2. $\gamma_{0}^{-1} \sim 3 $ ns.
3. $T \sim 300 K \sim 0.026$ eV.
4. $|\mathbf{r}_{i}-\mathbf{r}_{j}| \sim 10-100$ nm.

In the numerical calculation these values correspond to

1. $\omega_{0} = 1$.
2. $\gamma_{0} \sim 10^{-8}$.
3. $T \sim 0.015$.
4. $|\mathbf{r}_{i}-\mathbf{r}_{j}| \sim 0.01 - 0.1 \lambda_{0}$.




## Dual, triple, and quadraple time scales

The time scales are determined by the Bohr frequency $\omega_{0}$, dipole-dipole coupling strength $\Omega$, and spontaneous emission rate $\gamma_0$.  There are two cases:

1. $\omega_0 \gg \gamma_{0} > \Omega$
2. $\omega_0 \gg \Omega \gg \gamma_{0}$

Recall that $\Omega$ determines the time scale of energy exchange between the emitters. In the first case, the exchange time is slower than the decay process.  Hence, the is not enough time to make a transition and we can safely ignore the dipole-dipole coupling.   Then we have two time scale $\omega_{0}$ and $\gamma_{0}$  (dual time scales).  These time scales are typically at order of $\displaystyle\frac{\gamma_{0}}{\omega_{0}} = 10^{-8}$.  Since we are interested in the slow decay process, we will eliminate $\omega_{0}$ using the rotating frame.

When the distance between the emitters is less than $0.1 \lambda_{0}$, the dipole-dipole coupling gets significantly large.  Then, all three scales become important (triple time scales).   Typical orders are  $\displaystyle\frac{\Omega}{\omega_{0}}= 10^{-5}$ and $\displaystyle\frac{\gamma_{0}}{\omega_{0}}= 10^{-8}$.  Then, we need to eliminate both $\omega_{0}$ and $\Omega$, which can be done again with the rotating frame method but we also have to use the Davies-GKSL master equation.
