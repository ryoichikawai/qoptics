# Dimer IV: A Pair of quantum dots

In {numref}`ch:monomer-case2`, we have investigated coherent properties of light emitted by a quantum dot driven by non-resonant above-gap pumping.  Dephasing due to phonon scattering dominant role in the first-order coherence but the second-order coherence was not directly affected by the dephasing.  In this chapter, we investigate a pair of quantum dots driven by the same pumping method.   There are two key points to pay attention. 

1. How does dephasing affect the dipole-dipole coupling?
2. How does dephasing affect the HBT effect in the second-order coherence?

Recall that the second-order coherence for a signle QD is not affected by dephasing at all.  But that may change for $N=2$.
