# Photons and quantum radiation fields

## Classical radiation field

Classical radiation fields are described by electromagnetic waves. An electromagnetic wave is a solution to Maxwell's equations, and it represents a vector in an infinite-dimensional Hilbert space. Hence, the solution can always be expressed as a linear combination of basis functions $\{u_{i}(\mathbf{x})\}$:

$$
\mathbf{E}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \left( \mathcal{E}_{j} a_{j}(t) u_{j}(\mathbf{x}) +  \mathcal{E}^{*}_{j} a^{*}_{j}(t) u^{*}_{j}(\mathbf{x})\right )
$$(eq:field_decomposition)

where $\mathcal{E}_{j}$ and $\mathbf{e}_{j}$ are a constant with the dimension of the electric field and a polarization vector associated with the $j$-th basis function, respectively. The coefficients $a_{j}$ represent  a particular solution.  In optics, this basis is called an _optical mode_.

In free space,  the most popular choice of basis is plane waves, $u_{j}(\mathbf{x}) = e^{i\mathbf{k}_{j}\cdot \mathbf{x}}$. Correspondingly, we have $a_{j}(t) = a_{j} e^{-i \omega_{j} t}$ and $\mathcal{E}_{j} = i \sqrt{\frac{\hbar \omega_j}{2 \varepsilon_{0} V}}$. We have used discrete wave vectors in order to avoid mathematical complications but in the real world, we must use a continuous spectrum. 

In classical theory, the coefficients $a_{j}(t)$ are treated as stochastic variables. This randomness accounts for the apparent loss of coherence.  While this approach works quite well to reproduce experimental results, it is not based on the first principles.


For mathematical convenience, we use the complex expression $\mathbf{E} = \mathbf{E}^{(+)} + \mathbf{E}^{(-)}$  defined by

$$
\mathbf{E}^{(+)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}_{j} a_{j}(t) u_{j}(\mathbf{x}), \qquad \mathbf{E}^{(-)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}^{*}_{j} a^{*}_{j}(t) u^{*}_{j}(\mathbf{x}).
$$(eq:classical-fields)

and the intensity of the light is defined by

$$
I(\mathbf{x}) = \langle \mathbf{E}^{(-)}(\mathbf{x},t) \cdot \mathbf{E}^{(+)}(\mathbf{x},t)\rangle = \sum_{i,j} \mathbf{e}_{i}\cdot\mathbf{e}_{j}\, \mathcal{E}_{i} \mathcal{E}^{*}_{j} \, u_{i}(\mathbf{x}) u^{*}_{j}(\mathbf{x})  \langle a^{*}_{i}(t) a_j(t)\rangle
$$(eq:classical-intensity)

where $\langle \cdots \rangle$ denotes an ensemble average over the probability distribution of the random phase. This ensemble average is usually replaced with a temporal average in practical applications.

## Field quantization and photons

Quantizing radiation field {eq}`eq:classical-fields` is straightforward. The coefficients $a_{j}$ and $a^{*}_{j}$ become the annihilation operator $\hat{a}_{j}$ and the creation operator $\hat{a}^{\dagger}_{j}$ of a photon, respectively. They satisfy the usual commutation relation $[\hat{a}_{j},\hat{a}^{\dagger}_{j}] = 1$ and follow the usual operator algebra of a harmonic oscillator.  When the basis functions $u_{i}(\mathbf{x})$ and $u_{j}(\mathbf{x})$ are orthogonal, $[\hat{a}_{i},\hat{a}^{\dagger}_{j}] = 0$.  In real experimental condition, the basis functions may not be orthogonal, then the commutation does not vanishes.  Such cases will be discussed later.  Until then, we assume that the modes are orthogonal to each other.

We define photons using these creation and annihilation operators. Defining the vacuum state by $a_{j}|0_{j}\rangle=0$, we create a photon in  $j$-th mode as $|1_{j}\rangle = a^{\dagger}_{j}|0_{j}\rangle$. The operator $n_{j} = a^\dagger_{j}a_{j}$ is the number operator and $\langle n_{j} |a^\dagger_{j}a_{j} |n_{j}\rangle = n_{j}$ is the number of photons in the $j$-th mode.   A general state of photons in the mode is given by a superposition $|\phi_{j}\rangle = \sum_{n_j} c_{n_j} |n_{j}\rangle$.
For the system with multiple modes,  the state of photons is a superposition of $|\Phi\rangle = \sum_{n_1,n_2,\cdots} c_{n_1,n_2,\cdots} |n_1, n_2, \cdots \rangle$.


We can say that a photon is created or destroyed, and we can also count photons. That is all we can say about photons.  It is very important to keep in our mind that each photon is associated with a given mode.  A state $|1\rangle$ does not mean that there is one photon in the whole space. Until a mode is specified, $|1\rangle$ does not make a sense.


The free Hamiltonian of the radiation field is given by $\sum_{j} \frac{\omega_{j}}{2} \left(a^{\dagger}_{j} a_{j} + \frac{1}{2}\right)$ where $\omega_{j}$ is the energy of a photon in the $j$-th mode.

In the relativistic world, the number of photons is not invariant under a Lorentz transformation.  Hence, it is difficult to consider them as particles.   Some people even call the photon a pseudo particle.  This is why Glauber said "a photon is what a photodetector detects", which implies that we should not think about photons traveling around in the space. Ultimately, we consider the quantum electromagnetic fields to be fundamental whereas photons are emergent. 

When an electron in an emitter jumps from a higher energy state to a lower energy state, we say a photon is emitted.  What this statement mean is that the electron deposits a quantum of energy into the field, and then the energy stored in the field somehow travels to the detector and triggers a click. That is all we need for now. Photons reveal their existence only when the energy quanta is emitted to the field or absorbed from the field.

## Quantized radiation field

Here we consider the radiation field in a free space. It is natural to expand the field with the plane wave basis set. Then, the electric field at position $\mathbf{x}$ and time $t$ is given by

$$
\mathbf{E}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \left( \mathcal{E}_{j} a_{j} e^{-i (\omega_{j}t -  \mathbf{k}_{j} \cdot \mathbf{x}) } +  \mathcal{E}^{*}_{\ell} a^{\dagger}_{j} e^{i (\omega_{j}t - \mathbf{k}_{j} \cdot \mathbf{x})}\right )
$$(eq:quantum-free-field)

where $\mathcal{E}_{\ell} = i \sqrt{\frac{\hbar \omega_{\ell}}{2 \epsilon_{0} V}}$ and $\epsilon_{\ell}$ is the polarization vector for the $\ell$-th mode. The creation  and annihilation operators are in the Schrödinger picture. While it is a mathematically legitimate  solution, the solution {eq}`eq:quantum-free-field` has an issue with its physical interpretation.   The term with $e^{-i \omega_j t}$ propagates to the future, which make a sense,  but the other term with $e^{+i \omega_{j} t}$ propagates in the opposite direction of time.  Such a backward propagation is unphysical and usually ignored.

In order to avoid the difficulty, we splits the field to the two parts, the retarded field $\mathbf{E}^{(+)}(\mathbf{x},t)$ and the advanced field $\mathbf{E}^{(-)}(\mathbf{x},t)$:

$$
\mathbf{E}^{(+)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}_{j} a_{j} e^{-i(\omega_{j} t - \mathbf{k}_{\ell} \cdot \mathbf{x})}, \qquad
\mathbf{E}^{(-)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}^{*}_{j} a^{\dagger}_{j} e^{i (\omega_{j} t - \mathbf{k}_{\ell} \cdot \mathbf{x})}.
$$(eq:quantum-free-field-pm)

Remember that the electric field $\mathbf{E}$ is a quantum mechanical operator.  If the Heisenberg picture is used, $a$ and $a^\dagger$ are time-dependent and thus $\mathbf{E}$ is time-dependent.  If the Schrödinger picture is used, then $\mathbf{E}$ does not depend on time.  We will choose a convenient picture depending on the situation.

In theoretical calculations, we usually use the Heisenberg picture and thus the operators absorbe the time as $a_{j}(t) = a_{j} e^{-i \omega_{j} t}$ and $a^{\dagger}_{j}(t) = a^{\dagger}_{j} e^{i \omega_{j} t}$. The quantum expectation value of an observable $\hat{A}$ is defined as $\langle A(t) \rangle = \text{tr}(A(t) \rho)$ in the Heisenberg picture or $\text{tr}(A \rho(t))$ in the Schrödinger picture, where $\rho$ is a density operator. 


The mean intensity of light is given by

$$
I(\mathbf{x},t) = \langle \mathbf{E}^{(-)} \cdot\mathbf{E}^{(+)}\rangle = \sum_{i,j} \mathbf{e}_{i}\cdot\mathbf{e}_{j}\, \mathcal{E}^{*}_{i} \mathcal{E}_{j} \, e^{-i \mathbf{k}_{i}\cdot \mathbf{x}} e^{i \mathbf{k}_{j} \cdot \mathbf{x}}  \langle a^{\dagger}_{i}(t) a_{j}(t)\rangle
$$(eq:radiation-intensity)

This averaging replaces the ensemble average appeared in the classical intensity {eq}`eq:classical-intensity`.

In classical optics, the electric field is a measurable quantity.  In quantum field theory, the observation of electric field is quite different.  If the state of radiation field is an eigenstate of the number operator, the expectation value of the field vanishes, .i.e. $\langle \mathbf{E} \rangle=0$.  However, $\langle \mathbf{E}^2 \rangle > 0$ even when there is no photon. This is just the properties of a harmonic oscillator.

The classical-like electric field can be seen when the quantum state of the radiation is a _coherent state_, a mixture of different numbers of photons. The state of laser beam is an example. 

We also need to be careful about  $\mathbf{E}^{(\pm)}$.  They are not self-adjoint and thus not physical observables.  Measuring $\mathbf{E}^{(\pm)}$ is not possible.



## Measurement of radiation field

For massive particles such as electrons, we can measure many physical quantities, position, momentum, energy, .....

In classical optics, the intensity of light is often defined as 

$$
\begin{aligned}
I &= \langle \mathbf{E}\cdot \mathbf{E} \rangle \nonumber \\
&= 
\langle (\mathbf{E}^{(+)})^2 \rangle + \langle (\mathbf{E}^{(-)})^2 \rangle + \langle \mathbf{E}^{(+)}\cdot \mathbf{E}^{(-)} \rangle +  \langle \mathbf{E}^{(-)}\cdot \mathbf{E}^{(+)} \rangle.
\end{aligned}
$$(eq:intensity-fullterms)

The intensity defined by Eq. {eq}`eq:radiation-intensity` consists of only one of the four terms in Eq. 
{eq}`eq:intensity-fullterms`.  Are we missing some energy?  There are several reasons quantum optics uses Eq. {eq}`eq:radiation-intensity`.

**Infinite zero point energy**

Suppose that we have a vacuum state (no photon).  $\langle 0 | \mathbf{E}^{(+)}\cdot \mathbf{E}^{(-)} | 0  \rangle$ diverges due to the zero point energy of the infinite number of oscillators.  We want to remove this energy from the intensity of the radiation.

**Rapid oscillations**

Since $\mathbf{E}^{(+)} \propto e^{-i \omega t}$ and $\mathbf{E}^{(-)} \propto e^{i \omega t}$, the square terms in Eq. {eq}`eq:intensity-fullterms` are rapidly oscillating and the standard detector does not have sufficient time resolution to see the oscillation.  The detector averages out the oscillation and thus these terms vanish.  Since the actual detector can't see them, there is no reason to include them. This is just a practical reason. 

**What a detector sees**

A single photon detector records the arrival of individual photons.  Inside the detector, an electron gains energy from a photon by absorbing it.  Mathematically, this process is represented by $a^{\dagger}a$.  Only $\langle \mathbf{E}^{(-)}\cdot \mathbf{E}^{(+)} \rangle$ corresponds to the process in the detector.  The detail description of the detector is discussed in {numref}`sec:detector`.

We need to be careful about  $\mathbf{E}^{(\pm)}$.  They are not self-adjoint and thus not physical observable.  Measuring $\mathbf{E}^{(\pm)}$ is not possible.  Any observable quantity must be a product of the same number of $\mathbf{E}^{(+)}$ and $\mathbf{E}^{(-)}$. 

In classical optics, the electric field is a measurable quantity.  In quantum field theory, the observation of electric field is quite different.  If the state of radiation field is an eigenstate of the photon number operator, the expectation value of the field vanishes, i.e., $\langle \mathbf{E}\rangle=0$  despite $\langle \mathbf{E}^2 \rangle$ diverges even when there is no photon as we discussed above.

The classical-like electric field can be seen when photons are in a _coherent state_, a superposition of different photon numbers. The state of laser beam is an example.  In this lecture note, we are dealing with a weak radiation field so that we can count the number of photons arriving  at the detector.


## Measurement of photons

The measurement of a radiation field shown above ends up in the measurement of photons in relevant modes.  Actually photodetectors detect only the absorption of photons.  When a single photon arrived, it clicks.  From the recorded clicks, we can evaluate the intensity of the field, which we discuss later.  At the fundamental level, the mechanism of photon measurement is still unknown.  There are many issues which cannot be resolved with the current quantum mechanics and quantum field theory.  We will not discucss the profound issues here.

Since the photons are quantum object, it must follow the postulates of quantum measurement.  Since the quantum theory of photon measurement is still under development, we just apply the current understanding.  When a photon is detected, it is absorbed by the detector.  If the detector is monitoring a mode $j$, the number of photons in the mode decreases by one.  The detection of photons is stochastic and we can only talk about the detection rate or a probability that a photon is detected during a given time window.


For practical appications, it is important to identify the modes  photons are occupying.  To make story concrete, we consider a measurement of photons coming from two emitters.
