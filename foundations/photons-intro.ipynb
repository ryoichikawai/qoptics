# Photons and quantum radiation fields

Classical light is described by electromagnetic fields that are continuous both in space and time. Quantum light is also described by electromagnetic fields, but they are quantum fields. The quantum fields are also continuous. So, where is the "quantumness" in quantum light? Photons immediately pop up in our heads. But what are they? Unfortunately, they are poorly defined in the literature and misused by many people.

It is often said that Planck envented photon but is misleading. He did not quantize the radiation. In his theory the discreteness in the atom was responsible for the discrete energy the radiation.  It was Einstein who quantized the radiation in the theory of photoelectric effect. 

The word "photon" was coined by G. N. Lewis in 1926. {cite:p}`Lewis1926` In his words,

>It would seem inappropriate to speak of one of these hypothetical entities as a particle of light, a corpuscle of light, a light quantum, or a light quant, if we are to assume that it spends only a minute fraction of its existence as a carrier of radiant energy, while the rest of the time it remains as an important structural element within the atom.  It would also cause confusion to call it merely a quantum, for later it will be necessary to distinguish between the number of these entities present in an atom and the so-called quantum number.  I therefore take the liberty of proposing for this hypothetical new atom which is not light but plays an essential part in every process of radiation, the name _photon_.

He was clearly struggling to define what photon is.

In 1951, Albert Einstein said: 

> All these fifty years of conscious brooding have brought me no nearer to the answer to the question 'What are light quanta?' Nowadays every Tom, Dick and Harry thinks he knows it, but he is mistaken."{cite:p}`French1979`

Since the correct use of the photon concept is so difficult, W. E. Lamb suggested that 

> A license be required for use of the word photon.{cite:p}`Lamb1995`

Roy J. Glauber, who was given a license, defined the photon by stating:

> A photon is what a photodetector detects and also A photon is where the photodetector detects it

rather than defining a photon as a pre-existing "particle" in space. When four Nobel laureates in quantum optics spoke like these, the meaning of photons must be very difficult for laymen to understand.

We should not imagine that a photon travels along a trajectory like a classical particle. The position of a photon does not have a clear meaning. No position operator can even be defined for a photon, and thus we cannot measure its position. Recall that for another quantum particle, the electron, the position can be measured. So, the situation of the photon is quite different from that of other quantum particles. Nevertheless, we assume that the speed of a photon is the same as the speed of classical light $c$.[^1] 

Although we don't have a license, we shall use the word "photon" in these lecture notes, but only within a limited scope. We use the word in two instances: we say "A photon is emitted" when a quantum of energy is transferred from the emitter to the radiation field, and "A photon is detected" when the detector clicks (received a quantum of energy from the field).
