# Classical radiation field


A classical radiation field is described by an electromagnetic wave, which is a solution to the Maxwell's equations.  The solution is a vector in an infinite-dimensional Hilbert space and thus it can always be expressed as a linear combination of basis functions $\{u_{i}(\mathbf{x})\}$ at a given time:

$$
\mathbf{E}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \left( \mathcal{E}_{j} a_{j}(t) u_{j}(\mathbf{x}) +  \mathcal{E}^{*}_{j} a^{*}_{j}(t) u^{*}_{j}(\mathbf{x})\right )
$$(eq:field_decomposition)

where $\mathcal{E}_{j}$ and $\mathbf{e}_{j}$ are a constant with the dimension of the electric field and a polarization vector associated with the $j$-th basis function, respectively. The coefficients $a_{j}(t)$ are determined by the Maxwell equations.  In optics, the basis function is called an _optical mode_.

In free space,  the most popular choice of basis is plane waves, $u_{j}(\mathbf{x}) = e^{i\mathbf{k}_{j}\cdot \mathbf{x}}$. Correspondingly, we have $a_{j}(t) = a_{j} e^{-i \omega_{j} t}$ and $\mathcal{E}_{j} = i \sqrt{\frac{\hbar \omega_j}{2 \varepsilon_{0} V}}$ where $\omega_{j}$ is the frequency of the light and $\mathbf{k}_{j}$ is the corresponding wave vector. We have used a discrete spectrum in order to avoid mathematical complications but in the real world, we must use a continuous spectrum. 

In classical theory, the coefficients $a_{j}(t)$ are treated as stochastic variables, more precisely the coefficient contains a random phase factor $e^{i \phi(t)}$ where $\phi(t)$ is a random phase. This randomness accounts for the apparent loss of coherence.  While this approach works quite well to reproduce experimental results, it is not based on the first principles.

Mathematically, Eq. {eq}`eq:field_decomposition` is a valid solution to the Maxwell equations.  However, it contains a component that is not compatible with our current understanding of physics.  The terms with $e^{i \omega_{j}$ propagates in the negative direction of time. Such a backward propagation is unphysical and usually ignored.

In order to avoid the difficulty, we splits the field to the two parts, the retarded field $\mathbf{E}^{(+)}(\mathbf{x},t)$ and the advanced field $\mathbf{E}^{(-)}(\mathbf{x},t)$:

$$
\mathbf{E}^{(+)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}_{j} a_{j}(t) u_{j}(\mathbf{x}), \qquad \mathbf{E}^{(-)}(\mathbf{x},t) = \sum_{j} \mathbf{e}_{j} \mathcal{E}^{*}_{j} a^{*}_{j}(t) u^{*}_{j}(\mathbf{x}).
$$(eq:classical-fields)

and the intensity of the light is defined by

$$
I(\mathbf{x}) = \langle \mathbf{E}^{(-)}(\mathbf{x},t) \cdot \mathbf{E}^{(+)}(\mathbf{x},t)\rangle = \sum_{i,j} \mathbf{e}_{i}\cdot\mathbf{e}_{j}\, \mathcal{E}_{i} \mathcal{E}^{*}_{j} \, u_{i}(\mathbf{x}) u^{*}_{j}(\mathbf{x})  \langle a^{*}_{i}(t) a_j(t)\rangle
$$(eq:classical-intensity)

where $\langle \cdots \rangle$ denotes an ensemble average over the probability distribution of the random phase. This ensemble average is usually replaced with a temporal average in practical applications.
