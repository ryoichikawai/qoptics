# Monomer Case II: Model

We build a quantum master equation with two environments. Since $T \ll \omega_{0}$, we assume $T=0$.

## Quantum master equation

The Hamiltonian of the system is simply

$$
H_{0} = \frac{\sigma_{0}}{2} \sigma^{z}.
$$

For photon environment, we have one dissipator corresponding to spontaneous emission

$$
\mathcal{D}^{\downarrow}(\rho) = \gamma_{0} \left(\sigma^{-}\rho\sigma^{+} - \frac{1}{2} \left\{ \sigma^{+}\sigma^{-},\rho \right\} \right).
$$

The effective excitation discussed in the previous page is modeled as

$$
\mathcal{D}^{\uparrow}(\rho) = P \left(\sigma^{+}\rho\sigma^{-} - \frac{1}{2} \left\{ \sigma^{-}\sigma^{+},\rho\right\} \right)
$$

The dephasing due to phonon scattering can be expressed as

$$
\mathcal{D}^{\phi} (\rho) = - \frac{\gamma_{\phi}}{2} \left ( \sigma^{z} \rho \sigma^{z} - \frac{1}{2} \left\{\sigma^{z}\sigma^{z},\rho\right\} \right)
$$

The time scales of typical experimental setups are $\omega_{0} > \gamma_{\phi} \gg P > \gamma_{0}$.

First, we eliminate the time scale of $\omega_{0}$ by moving to a rotating frame using the unitary transformation $U - ^{-i H_{0} t}$.  The resulting quantum master equation is

$$
\frac{d}{dt} \rho = \mathcal{D}^{\downarrow}(\rho) + \mathcal{D}^{\uparrow}(\rho) + \mathcal{D}^{\phi} (\rho)
$$(eq:qme-monomer-QD)

Since $\frac{\gamma_{\phi}}{\gamma_{0}} \sim 10^{3}$, the direct integration of the master equation is not efficient.  Since the dephasing dissipator commutes with $H_{0}$, it does not acts on the diagonal element of the density operator.  Let us write down the equations of motion for each matrix elements.

$$
\begin{aligned}
\frac{d}{dt} \rho_{ee} &= P \rho_{gg} - \gamma_{0} \rho_{ee} \\
\frac{d}{dt} \rho_{eg} &= - \tfrac12\left(P + \gamma_{0} + 2\gamma_{\phi}\right) \rho_{eg} \\
\rho_{gg} &= 1 - \rho_{ee}
\end{aligned}
$$(eq:eom-monomer-QD)

Hence, the coherence term $\rho_{eg}$ is independent from the diagonal term and  vanishes very quickly.  The equation can be easily solved and the results is

$$
\begin{aligned}
\rho_{ee}(t) &= \rho_{ee}(0) e^{-(P+\gamma_{0}) t} + \frac{P}{P+\gamma_{0}} \left (1 - e^{-(P+\gamma_{0}) t} \right) \\
\rho_{eg}(t) &= \rho_{eg}(0) e^{- \tfrac12\left(P + \gamma_{0} + 2\gamma_{\phi}\right) t}
\end{aligned}
$$(eq:p_ee-monomer-QD)

The steady state is 

$$
\rho^\text{ss}_{ee} = \frac{P}{P+\gamma_{0}}, \qquad 
\rho^\text{ss}_{gg} = \frac{\gamma_{0}}{P+\gamma_{0}} 
$$(eq:rho_ss-monomer-QD)


## Coherence functions

Using the formula {eq}`eq:example-g1-explicit`, we obtain

$$
g^{(1)}(\tau) = e^{-\frac{1}{2}(P + \gamma_{0}+2 \gamma_{\phi})t} \approx e^{-\gamma_{\phi}t}
$$(eq:g1-monomer-QD)

Similarly using {eq}`eq:example-g2-explicit` we obtain the second-order coherence function

$$
g^{(2)}(\tau) = 1 - e^{-(P + \gamma_{0})\tau}
$$

which does not involve $\gamma_{\phi}$.  

```{admonition} Effect of dephasing
class: important
:name: adm:dephasig
While the dephasing dominates in the first-order coherence, the second-order coherence is not affected by the dephasing.
```
