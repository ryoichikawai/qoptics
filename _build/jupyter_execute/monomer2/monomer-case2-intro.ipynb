# Monomer Case II:   Pumping Quantum Dot

In previous model, we discussed resonant or near resonant fluorescence in which the emitter was directly excited by a laser beam.  The driving itself was not dissipative.  In this chapter, we discuss another driving method often used to pump quantum dots.   A laser excites an electron to the conduction band leaving a hole.  The electron is quickly deexcited by emitting phonons and forms an exciton.  The exciton emits a photon through recombination. See {numref}`fig:QD-pumping`.  The electron does not have to bind to the hole just created with the electron. If it binds to another hole, the exciton is said to be non-geminate.



```{figure} pumping.webp
:name: fig:QD-pumping

Laser pumping to form an exciton through the conduction band.  The dotted arrow indicates an effective excitation used in the quantum master equation.

Since this process involves phonon emission, it is dissipative.  Unlike the resonant fluorescence, we cannot describe the excitation as unitary evolution unless the phonons are explicitly included in the calculation.  In our model, the pumping is dissipative much like thermal excitation.  However, we have a different environment, a phonon bath.  The phonon bath not only contribute the formation of the exciton. Phonon scattering also causes dephasing.  Due to the presence of two environments, phonon gas and photon gas, the quantum master equation becomes a little more complicated.  The goal of this chapter is to learn how to deal with the two baths, in particular time scales of individual processes.  

The pumping rate $P$ is determined by the laser intensity.  It should be fast enough to create an exciton, thus $P > \gamma$.  We assume $P = \gamma \sim 10 \gamma$.  The non-radiative decay rate $\Gamma_{\phi}$ is expected to be very fast (below ps), which allows us to think the rate of exciton formation is essentially $P$.  Base of this observation, we will build a quantum master equation in the next section.

For $N=1$, this is not a problem as you see in the above discussion.  However, for $N>1$ we need additional approximation to separate it from other time scales.
