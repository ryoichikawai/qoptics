(sec:dimer-case2-theory)=
# Dimer II: Theory

## Equations of motion

The diagonal elements do not coupled to the off-diagonal elements and their equation of motions are

$$
\begin{aligned}
\frac{d}{dt}\rho_{gg} &= - 2 \Gamma_{11} N_{T} \rho_{gg} + \Gamma^{+}(N_{T}+1) \rho_{ss} + \Gamma^{-} (N_{T}+1) \rho_{aa}\\
\frac{d}{dt}\rho_{aa} &= - \Gamma^{-}(2 N_{T}+1) \rho_{aa} + \Gamma^{-} N_{T} \rho_{gg} + \Gamma^{-} (N_{T}+1) \rho_{ee} \\
\frac{d}{dt}\rho_{ss} &= - \Gamma^{+}(2 N_{T}+1) \rho_{ss} + \Gamma^{+} N_{T} \rho_{gg} + \Gamma^{+} (N_{T}+1) \rho_{ee} \\
\frac{d}{dt}\rho_{ee} &= - 2 \Gamma_{11} (N_{T}+1) \rho_{ee} + \Gamma^{-} N_{T} \rho_{aa} + \Gamma^{+} N_{T} \rho_{ss}
\end{aligned}
$$

where $\Gamma^{\pm} = \Gamma_{11}\pm\Gamma_{12}$.  One equation can be eliminated using the normlization condition $\rho_{gg}+\rho_{aa}+\rho_{ss}+\rho_{ee}=1$.

$$
\frac{d}{dt} \begin{pmatrix} \rho_{gg} \\ \rho_{aa} \\ \rho_{ss} \\ \rho_{ee} \end{pmatrix}
= 
\begin{pmatrix}
- 2 \Gamma_{11} N_{T} & \Gamma^{-} (N_{T}+1) & \Gamma^{+}(N_{T}+1) & 0 \\
\Gamma^{-} N_{T} &  - \Gamma^{-}(2 N_{T}+1) & 0 & \Gamma^{-} (N_{T}+1) \\
\Gamma^{+} N_{T} & 0 & - \Gamma^{+}(2 N_{T}+1) & \Gamma^{+} (N_{T}+1) \\
0 & \Gamma^{-} N_{T} & \Gamma^{+} N_{T} & - 2 \Gamma_{11} (N_{T}+1)
\end{pmatrix}
\begin{pmatrix} \rho_{gg} \\ \rho_{aa} \\ \rho_{ss} \\ \rho_{ee} \end{pmatrix}
$$


The steady state solution is 

$$
\begin{aligned}
\rho_{gg}^{ss} &= \frac{(1+N_{T})^2}{(1+2 N_{T})^2} = \frac{e^{\omega_{0}/T}}{(1+e^{-\omega_{0}/T})^2}\\
\rho_{ss}^{ss} &= \frac{(N_{T} (N_{T}+1)}{(1+2 N_{T})^2} = \frac{1}{(1+e^{-\omega_{0}/T})^2} \\
\rho_{aa}^{ss} &= \frac{(N_{T} (N_{T}+1)}{(1+2 N_{T})^2} = \frac{1}{(1+e^{-\omega_{0}/T})^2} \\
\rho_{ee}^{ss} &= \frac{N_{T}^2}{(1+2 N_{T})^2}  = \frac{e^{-\omega_{0}/T}}{(1+e^{-\omega_{0}/T})^2}
\end{aligned}
$$(eq:dimer-case2-steady-state)

which agree with the Boltzmann distribution.  Notice that neither $\Gamma_{11}$ nor $\Gamma_{12}$ appears in the steady state because the thermal equilibrium only depends on the energy eigenvalues.  

Thanks to the Dicke basis, the equations of motion for the off-diagonal elements are rather simple.  First, two elements are  self contained as

$$
\begin{aligned}
\frac{d}{dt} \rho_{as} &= - \Gamma_{11} (2 N_{T}+1) \rho_{as} \\
\frac{d}{dt} \rho_{ge} &= \left[ i 2\omega_{0} - \Gamma_{11}(2 N_{T}+1) \right] \rho_{ge}
\end{aligned}
$$

which can be solved right away

$$
\begin{aligned}
\rho_{as}(t) &= e^{-\Gamma_{11} (2 N_{T}+1) t} \rho_{as}(0) \\
\rho_{ge}(t) &= e^{i 2\omega_{0} t} e^{-\Gamma_{11} (2 N_{T}+1) t} \rho_{ge}(0)
\end{aligned}
$$

Both decay quickly at the rate of $\Gamma_{11} (2 N_{T}+1)$.  

The equations of motion for the off-diagonal elements inside the channel 1 are a pair of ODEs:

$$
\begin{aligned}
\frac{d}{dt} \rho_{gs} &= \left[ i \omega_{0} - \Gamma_{11} N_{T} - (\Gamma_{11}+\Gamma_{12}) \left(N_{T}+\frac{1}{2}\right) \right]\rho_{gs} + (\Gamma_{11}+\Gamma_{12})(N_{T}+1) \rho_{se}\\
\frac{d}{dt} \rho_{se} &= \left[ i \omega_{0} - \Gamma_{11} (N_{T}+1) - (\Gamma_{11}+\Gamma_{12}) \left(N_{T}+\frac{1}{2}\right) \right] \rho_{se} + (\Gamma_{11}+\Gamma_{12})N_{T} \rho_{gs}
\end{aligned}
$$

It is easy to find that $\rho_{gs}$ and $\rho_{se}$ vanishes as $t \rightarrow \infty$.

Similarly for channel 2, we have a coupled ODE:

$$
\begin{aligned}
\frac{d}{dt} \rho_{ga} &= \left[ i \omega_{0} - \Gamma_{11} N_{T} - (\Gamma_{11}-\Gamma_{12}) \left(N_{T}+\frac{1}{2}\right) \right]\rho_{ga} + (\Gamma_{11}-\Gamma_{12})(N_{T}+1) \rho_{ae}\\
\frac{d}{dt} \rho_{ae} &= \left[ i \omega_{0} - \Gamma_{11} (N_{T}+1) - (\Gamma_{11}-\Gamma_{12}) \left(N_{T}+\frac{1}{2}\right) \right] \rho_{ae} + (\Gamma_{11}-\Gamma_{12})N_{T} \rho_{ga}
\end{aligned}
$$

and $\rho_{ga}$ and $\rho_{ae}$ vanishes as $t \rightarrow \infty$.

Since all off-diagonal elements vanish in the steady state, there is no coherence between the Dicke basis states.

(ssec:broken-ergodicity)=
## Broken Ergodicity


Recall that we are taking limit $\displaystyle\lim_{\Gamma_{12} \rightarrow \gamma_0}$.  At the same time, we tale another limit $\displaystyle\lim_{t \rightarrow \infty}$ to get the steady state.  These two limits do not commute, that is $\displaystyle\lim_{\Gamma_{12} \rightarrow \gamma_0} \lim_{t \rightarrow \infty} \rho \ne \lim_{t \rightarrow \infty} \lim_{\Gamma_{12} \rightarrow \gamma_0} \rho$. The steady state {eq}`eq:dimer-case2-steady-state` is 
obtained by $\displaystyle\lim_{\Gamma_{12} \rightarrow \gamma_0} \lim_{t \rightarrow \infty} \rho$.  
If we take the limit $\displaystyle\lim_{\Gamma_{12} \rightarrow \gamma_0}$ first, the anti-symmetric state $|a\rangle$ is completely isolated from the radiation field.  Furthremore, it is an eigenstate of the Hamiltonian.  There is no process that changes its population. Because of this isolation,  $\rho_{aa}^{ss} = \rho_{ss}(0)$ and the system does not reach the thermal state {eq}`eq:dimer-case2-steady-state`.  The steady state depends on the initial condition!  In thermodynamics, this issue is known as broken ergodicity. 

When the ergodicity is broken, we need to remove $|a\rangle$ from the Hilbert space.  Then, the steady state should be 

$$
\begin{aligned}
\rho_{gg}^{ss} &= \frac{(1+N_{T})^2}{3 N_{T}^2 + 3 N_{T}+1} = \frac{e^{\omega_{0}/T}}{1 + e^{-\omega_{0}/T} + e^{+\omega_{0}/T}}\\
\rho_{ss}^{ss} &= \frac{(N_{T} (N_{T}+1)}{3 N_{T}^2 + 3 N_{T}+1} = \frac{1}{1 + e^{-\omega_{0}/T} + e^{+\omega_{0}/T}} \\
\rho_{aa}^{ss} &= 0 \\
\rho_{ee}^{ss} &= \frac{N_{T}^2}{3 N_{T}^2 + 3 N_{T}+1}  = \frac{e^{-\omega_{0}/T}}{1 + e^{-\omega_{0}/T} + e^{+\omega_{0}/T}}
\end{aligned}
$$(eq:dimer-case2-steady-state-non-ergodic)

We will see that the asymptotic limit $\Gamma_{12} \rightarrow $\gamma_0$ does not agree with $\Gamma_{12}=\gamma_{0}$.

## Intensity of radiation

The intensity of radiation measured at $\mathbf{R}$ is given by Eq. {eq}`eq:intensity-R`.  For a dimer, it is

$$
\begin{aligned}
I(\mathbf{R},t) &= \Gamma(\mathbf{R})
\left[\langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_{ss} + \langle\sigma^{+}_{2}\sigma^{-}_{2}\rangle_{ss}
+ e^{i \phi } \langle\sigma^{+}_{1}\sigma^{-}_{2}\rangle_{ss} + e^{- i \phi} \langle\sigma^{+}_{2}\sigma^{-}_{1}\rangle_{ss} \right) \\
& \approx \Gamma(\mathbf{R}) \left[\langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_{ss} + \langle\sigma^{+}_{2}\sigma^{-}_{2}\rangle_{ss}
+ \left(1+i \phi - \frac{\phi^2}{2}\right)\langle\sigma^{+}_{1}\sigma^{-}_{2}\rangle_{ss} +  \left(1-i \phi - \frac{\phi^2}{2}\right)\langle\sigma^{+}_{2}\sigma^{-}_{1}\rangle_{ss}  \right] \\
&= \Gamma(\mathbf{R}) \left[ \left(2-\frac{\phi^2}{2}\right) \langle \eta^{+}_{1}\eta^{-}_{1}\rangle_{ss}  + \frac{\phi^2}{2}  \langle \eta^{+}_{2}\eta^{-}_{2}\rangle_{ss} + i \phi  \left(\langle \eta^{+}_{1}\eta^{-}_{2}\rangle_{ss} -\langle \eta^{+}_{2}\eta^{-}_{1}\rangle_{ss}\right) \right] \\
&= \Gamma(\mathbf{R}) \left[ \left(2-\frac{\phi^2}{2}\right) \langle \eta^{+}_{1}\eta^{-}_{1}\rangle_{ss}  + \frac{\phi^2}{2}  \langle \eta^{+}_{2}\eta^{-}_{2}\rangle_{ss} \right]
\end{aligned}
$$

where $\phi = k_{0}r_{12} \cos(\chi)$ with the angle $\chi$ between $\mathbf{R}$ and $\mathbf{r}_{12}$. Since we assume $r_{12} \ll \lambda_{0}$ ($\phi \ll 1$) we expand the phase factor up to $\phi^2$. The cross correlation $\langle \eta^{+}_{1} \eta^{-}_{2}\rangle_{ss}$ and $\langle \eta^{+}_{2} \eta^{-}_{1}\rangle_{ss}$ vanish since $\rho^{ss}_{as} = \rho^{ss}_{sa}=0$.
   $\Gamma(\mathbf{R}) = \left(\frac{\omega_{0}^{2}d}{c^2}\right)^2 \left [ 1 - \left(\hat{\mathbf{d}}\cdot\hat{\mathbf{R}}\right)^2\right]$.


Interestingly, when $\phi=0$ (either $r_{12}=0$ or $\theta=\pi/2$), $\eta^{\pm}_{2}$ (a-channel) does not contribute to the intensity.  If the detector can't distinguish individual emitters, only photons emitted by the s-channel is detected.  The radiation from the a-channel vanishes due to interference.

Now, evaluate the correlations

$$
\langle \eta^{+}_{1}\eta^{-}_{1}\rangle =  \rho^{ss}_{ss} + \rho^{ss}_{ee}, \qquad
\langle \eta^{+}_{2}\eta^{-}_{2}\rangle =  \rho^{aa}_{ss} + \rho^{ss}_{ee}
$$

We find the total intensity 

$$
I(\mathbf{R}) = \Gamma(\mathbf{R})\left[2 (\rho^{ss}_{ee}+\rho^{ss}_{ss}) + \frac{\phi^2}{2}\left(\rho^{ss}_{aa}-\rho^{ss}_{ss}\right)\right]
$$

When $\Gamma_{12} > 0$ (the system is ergodic), $\rho^{ss}_{ss}=\rho^{ss}_{aa}$ and thus $I(\mathbf{R}) = 2 \Gamma(\mathbf{R}) (\rho^{ss}_{ee}+\rho^{ss}_{ss})$ regardless of $\phi$.  Integrating over the the direction of $\mathbf{R}$, the net intensity emitted by the dimer is $I = 2 \gamma_{0} (\rho^{ss}_{ee}+\rho^{ss}_{ss})$ which is identical to the case where $r_{12} > \lambda_{0}$.  In terms of the radiation intensity, there is no superradiance for a dimer.

## Second-order coherence


$$
\begin{aligned}
G^{(2)}(\tau)/\Gamma^2(\mathbf{R}) &= (4-2 \phi^2) \langle \eta^{+}_{1}(0)\eta^{+}_{1}(\tau) \eta^{-}_{1}(\tau) \eta^{-}_{1}(0)\rangle \\
&+\phi^2 \left[\langle\eta^{+}_{1}(0)\eta^{+}_{2}(\tau)\eta^{-}_{2}(\tau)\eta^{-}_{1}(0)\rangle + \langle\eta^{+}_{2}(0)\eta^{+}_{1}(\tau)\eta^{-}_{1}(\tau)\eta^{-}_{2}(0)\rangle\right] \\
&-\phi^2 \left[\langle\eta^{+}_{1}(0)\eta^{+}_{1}(\tau)\eta^{-}_{2}(\tau)\eta^{-}_{2}(0)\rangle - \langle\eta^{+}_{1}(0)\eta^{+}_{2}(\tau)\eta^{-}_{1} (\tau)\eta^{-}_{2}(0)\rangle \right]\\
&+ \phi^2 \left[\langle\eta^{+}_{2}(0)\eta^{+}_{1}(\tau)\eta^{-}_{2}(\tau)\eta^{-}_{1}(0)\rangle - \langle\eta^{+}_{1}(0)\eta^{+}_{2}(\tau)\eta^{-}_{2}(\tau)\eta^{-}_{1}(0)\rangle\right] \\
&+2 i \phi \left[ \langle\eta^{+}_{1}(0)\eta^{+}_{1}(\tau)\eta^{-}_{1}(\tau)\eta^{-}_{2}(0)\rangle - \langle\eta^{+}_{2}(0)\eta^{+}_{1}(\tau)\eta^{-}_{1}(\tau)\eta^{-}_{1}(0)\rangle\right] \\
&+2 i \phi \left[ \langle\eta^{+}_{1}(0)\eta^{+}_{1}(\tau)\eta^{-}_{2}(\tau)\eta^{-}_{1}(0)\rangle - \langle\eta^{+}_{1}(0)\eta^{+}_{2}(\tau)\eta^{-}_{1}(\tau)\eta^{-}_{1}(0)\rangle\right]
\end{aligned}
$$

and the normalized second-order coherence function 

$$
g^{(2)}(\tau) = \frac{G^{(2)}(\tau)}{4\left(\rho^{ss}_{ee}+\rho^{ss}_{ss}\right)^2}
$$

The no-delay function is

$$
g^{(2)}(0) = \frac{(4-\phi^2)*\rho^{ss}_{ee}}{4\left(\rho^{ss}_{ee}+\rho^{ss}_{ss}\right)^2}
$$(eq:g2_0-dimer-case2)

For $r_{12} > 0$, we can substitute the steady state {eq}`eq:dimer-case2-steady-state` to Eq. {eq}`eq:g2_0-dimer-case2`.  The result is simply

$$
g^{(2)}(0)  = 1 - \frac{\phi^2}{4}
$$

which approaches to 1 as $\phi \rightarrow 0$.


For $r_{12}=0$ and thus $\phi=0$, we use the steady state {eq}`eq:dimer-case2-steady-state-non-ergodic` and obtain

$$
g^{(2)}(0) = 1 - \frac{q}{(1+q)^2}
$$

which is bound between $\frac{3}{4}$ and $1$.
where $q = e^{-\omega_{0}/T}$.
