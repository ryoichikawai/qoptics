# Dimer II: Coherence by indistinguishability

In {numref}`ch:dimer-case1`, we considered vanishing off-diagonal dissipators ($\Gamma_{12}=0$). In this chapter, we consider the opposite limit, $\Gamma_{12} \rightarrow \gamma_{0}$ where the distance between the two emitters is smaller than the wave length ($k_{0}r_{12} \ll 1$). We further assume that the dipole-dipole coupling is negligible ($\Omega_{12} \ll \omega_{0}$) and thus the energy spectrum is determined solely by the free Hamiltonian $H_{0}$.  Only the difference from {numref}`ch:dimer-case1` is that $\Gamma_{12}=\gamma_{0}$. 

There is an important issue to be discussed.  When $\Gamma_{12}=\gamma_{0}$, there is a dark state that does not interact with the radiation field. Any finite population in the dark state is preserved and hence the steady state depends on initial conditions.  In other words, the dark state does not thermalize. On the other hand, if $\Gamma_{12}$ is just infinitesimally close to $\gamma_{0}$, the state slowly thermalizes.  Hence, _ergodicity is broken_.  We need to treat the two cases separately, one with exactly $\Gamma_{12}=\gamma_{0}$ and the other with $\Gamma_{12} = \gamma_{0} + \delta^{+}$.

When $\Gamma_{12}=\gamma_{0}$, the situation matches to superradiance.  Based on the standard theory of superradiance,  we might expect that the intensity is proportional to $N^2$ where $N$ is the number of emitters.  Is the intensity doubled for $N=2$.  It won't happen for a dimer.  However, the signature of superradiance can be seen in the full counting statistics and also in the second-order coherence.
