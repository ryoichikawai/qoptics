# Dimer I: Incoherent emitters 

In {numref}`ch:coherence`, we discussed the origin of decoherence.  When the radiation field can't distinguish the emitters ($\langle \phi_{1} | \phi_{1}^\prime\rangle \approx 1$), coherence is observed in the subsystem.  Otherwise ($\langle \phi_{1} |\phi_{1}^\prime\rangle \ll 1$), we don't see coherence.   The field very close to the emitters (near field) can obviously see individual emitters.    However, the field far from the emitters (far field) may not be able to see them separately.  Recall that the far field is expressed by Eq.{eq}`eq:source_field_approx` in which the location of the emitters appears in the phase factor, $e^{- i k_{0} \hat{\mathbf{R}} \cdot \mathbf{r}_{12}}$.   Roughly speaking, if $r_{12} k_{0} \ll 1$, then the phase difference vanishes and the far field can't distinguish the emitters.  Otherwise, the detector can distinguish the individual emitters.  From the standpoint of the emitters, the same information is included in $\Gamma_{i\ne j}$. (See {eq}`eq:Gamma_ij`.)

In this chapter, we investigate the coherence properties of thermal fluorescence emitted from a pair of emitters (we shall call them a dimer) separated by a large distance $r_{12} k_{0} \gg 1$. 

See Ref. {cite:p}`Ficek2002` for a comprehensive review of two-atom systems.
