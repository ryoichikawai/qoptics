# Dimer I: Model

## System

Each emitter is modeled by a two-level system as defined in {numref}`sec:emitter`.  For computational purpose, we use the standard product basis set $|gg\rangle$, $|ge\rangle$, $|eg\rangle$, and $|ee\rangle$ (see {numref}`sec:emitter` for the definition of the product base.)

The four-dimensional spin operators are defined by (see {numref}`sec:emitter`)

$$
\sigma_{1}^{\mu} = \sigma^{\mu} \otimes I, \qquad \sigma_{2}^{\mu} = I \otimes \sigma^{\mu} , \qquad \mu = x,y,z, \pm.
$$(eq:dimer-spin-ops)

where $I$ is the identity operator in the two-dimensional Hilbert space.

The Hamiltonian of the free system is then expressed as

$$
H_{0} = \sum_{j=1}^{2} \frac{\omega_0}{2} \sigma^{z}_{j}
$$(eq:dimer-H0)

We assume that the dipole-dipole coupling is negligible.  Since the Hamiltonian is diagonal in the product basis, naturally we use it.  The Dicke basis vectors are also energy eigenvectors due to the degeneracy of $|eg\rangle $ and $|ge\rangle$.  Using the Dicke basis generates the same result as the product basis.  Here are the energy eigenvectors and the corresponding eigen energies.

$$
\begin{aligned}
|u_{0}\rangle &= |ee\rangle, \qquad && E_{0} = \omega_{0}\\
|u_{1}\rangle &= |eg\rangle, \qquad && E_{1} = 0\\
|u_{2}\rangle &= |ge\rangle, \qquad && E_{2} = 0\\
|u_{3}\rangle &= |gg\rangle, \qquad && E_{3} = -\omega_{0}
\end{aligned}
$$(eq:dimer-pbasis)

There is only one Bohr frequency $\omega_{0}$.

## Dissipator


As discussed in {numref}`sec:openqm-qme-optical`, $\Gamma_{12}$ vanishes when the distance between the two emitters is larger than the wave length.  Accordingly, the dissipator for the pair is given by

$$
\begin{aligned}
\mathcal{D}(\rho) 
&=  \frac{1}{2} \gamma^{\downarrow} \left(2 \sigma^{-}_{1} \rho \sigma^{+}_{1} - \sigma^{+}_{1}\sigma^{-}_{1} \rho - \rho \sigma^{+}_{1}\sigma^{-}_{1} \right) \\
&+  \frac{1}{2} \gamma^{\downarrow} \left(2 \sigma^{-}_{2} \rho \sigma^{+}_{2} - \sigma^{+}_{2}\sigma^{-}_{2} \rho - \rho \sigma^{+}_{2}\sigma^{-}_{2} \right) \\
&=  \frac{1}{2} \gamma^{\uparrow} \left(2 \sigma^{+}_{1} \rho \sigma^{-}_{1} - \sigma^{-}_{1}\sigma^{+}_{1} \rho - \rho \sigma^{-}_{1}\sigma^{+}_{1} \right) \\
&+  \frac{1}{2} \gamma^{\uparrow} \left(2 \sigma^{+}_{2} \rho \sigma^{-}_{2} - \sigma^{-}_{2}\sigma^{+}_{2} \rho - \rho \sigma^{-}_{2}\sigma^{+}_{2} \right)
\end{aligned}
$$(eq:dimer-case1-dissipator)

where the emission and absorption rates are defined $\gamma^{\downarrow} = \gamma_{0} (\bar{n}+1)$ and $\gamma^{\uparrow} = \gamma_{0} \bar{n}$.  See {numref}`sec:openqm-qme-example` for details.

The off-diagonal dissipator is omitted since $\Gamma_{12} \approx 0$ at $k_{0} r_{12} \gg 1$. 

There are two cascading channels.  Channel 1: $|u_{0}\rangle \Rightarrow |u_{1}\rangle \Rightarrow |u_{3}\rangle$ and Channel 2: $|u_{0}\rangle \Rightarrow |u_{2}\rangle \Rightarrow |u_{3}\rangle$.  Two channels are perfectly symmetric.

## Steady state

The state of the emitters reaches a steady state after $t \gamma_{0} \gg 1$.  The steady state should be a Gibbs state

$$
\rho_\text{ss} = \frac{1}{Z} \sum_{k} e^{-E_{k}/T} |u_{k}\rangle\langle u_{k}|
$$(eq:dimer-steadystate)

where $Z$ is the partition function.  Since the state is diagonal in the product basis, there is no coherence between $|eg\rangle$ and $|ge\rangle$.  Hence $\langle \sigma^{+}_{1}\sigma^{-}_{2}\rangle = 0$.

For simplicity, we write the steady state as $\rho_\text{ss} = \text{diag}\left(p^{ee},p^{eg},p^{ge},p^{gg}\right)$.  Since there is no interaction between the emitters, there is no correlation between them and thus the state is a product of individual states $\rho_\text{ss} = \rho_{1} \otimes \rho_{2}$ where $\rho_{j}$ is the state of the $j$-th emitter. The individual state is also diagonal, which we write as $\rho_{j} = \text{diag}(p^{e}_{j},p^{g}_{j})$ where

$$
\begin{aligned}
p^{e}_1 &= \sqrt{(p^{ee})^2 + (p^{eg})^2}, \qquad &&p^{g}_1 = \sqrt{(p^{ge})^2 + (p^{gg})^2}\\
p^{e}_2 &= \sqrt{(p^{ee})^2 + (p^{ge})^2}, \qquad &&p^{g}_2 = \sqrt{(p^{eg})^2 + (p^{gg})^2}.
\end{aligned}
$$(eq:dimer-state-facterization)



## Detection I

Let us assume that the detector can distinguish the sources of light.  Then, we have two detection operators:

$$
D_{j} (\hat{\mathbf{n}_{j}},t) =   e^{-i k_{0} \hat{\mathbf{n}_{j}} \cdot \mathbf{r}_{j}} \sigma^{-}_{j}(t),
$$(eq:dimer-detector-model1)

which depends on the location of the detector with respect to the center of the dimer. The unit vector $\hat{\mathbf{n}_j}$ denotes the direction from the center to the detector. Using the spherical cordinates $(\theta,\phi)$, $\hat{\mathbf{n}_j} = (\sin\theta_{j}\cos\phi_{j},\sin\theta_{j}\sin\phi_{j},\cos\theta_{j})$.  Since the two detection areas very close to each other we assume that $\mathbf{n}_{1} = \mathbf{n}_{2}$.

The net intensity is given by

$$
I(\hat{\mathbf{n}},t) = W(\hat{\mathbf{n}}) \sum_{j} \langle D^{\dagger}_{j}(\mathbf{n},t) D_{j}(\hat{\mathbf{n}},t) \rangle = W(\hat{\mathbf{n}}) \sum_{j} \langle\sigma^{+}_{j}(t) \sigma^{-}_{j}(t)\rangle
$$(eq:dimer-intensity-model1)

with 

$$
W(\hat{\mathbf{n}}) \equiv \frac{3\gamma_0}{8\pi}\left[1 - \left(\hat{\mathbf{n}}\cdot\hat{\mathbf{d}}\right)^2\right] = \frac{3\gamma_0}{8\pi} \sin^2\theta
$$(eq:dimer-intensity-weight)

where $\theta$ is the angle between $\hat{\mathbf{n}}$ and $\hat{\mathbf{d}}$.  Obviously, the intensity vanishes if the dipole moment points to the detector.


## Detection II

It is possible to navigate photons from the two different sources to a single detection zone even two sources are separated by a large distance.  As long as detection zone is sufficiently narrow so that the phase difference of two beams remains constant, they are coherent.  Then, we have only one detection operator

$$
D = \sum_{j} D_{j} = \sum_{j} e^{-i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{j}} \sigma^{-}_{j}(t)
$$(eq:dimer-detector-model2)

Taking into account the lack of coherence between $|u_{1}\rangle$ and $|u_{2}\rangle$, we obtain the net intensity

$$
I(\hat{\mathbf{n}},t) =  W(\hat{\mathbf{n}}) \langle D^{\dagger}(\hat{\mathbf{n}},t) D(\hat{\mathbf{n}},t) \rangle = W(\hat{\mathbf{n}}) \sum_{ij} e^{-i k_{0} \hat{\mathbf{n}} \cdot \left(\mathbf{r}_{i} - \mathbf{r}_{j}\right)} \langle \sigma^{+}_{i}\sigma^{-}_{j}\rangle = W(\hat{\mathbf{n}}) \sum_{j}  \langle \sigma^{+}_{j}(t)\sigma^{-}_{j}(t)\rangle
$$(eq:dimer-intensity-model2)

which is the same as the intensity obtained by the detector model I.  However, two models will exhibit different coherence properties.

## First-order coherence function

Using the detector model I,  the first-order correlation is given by

$$
G^{(1)}(\hat{\mathbf{n}}_{1},t_1;\hat{\mathbf{n}}_{2},t_2) = \sum_{j} \langle D_{j}^{\dagger}(\hat{\mathbf{n}}_{1},t_1) D_{j}(\hat{\mathbf{n}}_{2},t_2)\rangle_\text{ss}
$$(eq:dimer-G1-model1)

We are particular interested in the temporal correlation measured at a single point.  Then, the normalized first-order coherence function is

$$
g^{(1)}(\hat{\mathbf{n}};t+\tau,t) = \frac{\sum_{j} \langle D_{j}^{\dagger}(\hat{\mathbf{n}},t+\tau) D_{j}(\hat{\mathbf{n}},t)\rangle_\text{ss}}{\sum_{j}I_{j}(\hat{\mathbf{n}},t)}
= \frac{\sum_{j} \langle\sigma^{+}_{j}(t+\tau) \sigma^{-}_{j}(t)\rangle_\text{ss}}{\sum_{j} \langle\sigma^{+}_{j}(t) \sigma^{-}_{j}(t)\rangle_\text{ss}}
= \frac{\sum_{j} \text{tr} \left[ \sigma^{-}_{j} e^{\mathcal{L}\tau} \left(\rho_\text{ss} \sigma^{+}_{j}\right) \right]}{\sum_{j}\text{tr}\left[\sigma^{+}_{j}\sigma^{-}_{j}\rho_\text{ss}\right]}
$$(dimer-g1-model1)

Similarly using the detector model II,

$$
G^{(1)}(\hat{\mathbf{n}}_{1},t_1;\hat{\mathbf{n}}_{2},t_2) = \langle D^{\dagger}(\hat{\mathbf{n}}_{1},t_1) D(\hat{\mathbf{n}}_{2},t_2)\rangle_\text{ss}
$$(eq:dimer-G1-model2)

and 

$$
g^{(1)}(\hat{\mathbf{n}};t+\tau,t) = \frac{\langle D^{\dagger}(\hat{\mathbf{n}},t+\tau) D(\hat{\mathbf{n}},t)\rangle_\text{ss}}{I(\hat{\mathbf{n}},t)}
= \frac{\sum_{ij} \langle\sigma^{+}_{i}(t+\tau) \sigma^{-}_{j}(t)\rangle_\text{ss}}{\sum_{j} \langle\sigma^{+}_{j}(t) \sigma^{-}_{j}(t)\rangle_\text{ss}}
= \frac{\sum_{ij} \text{tr} \left[ \sigma^{-}_{j} e^{\mathcal{L}\tau} \left(\rho_\text{ss} \sigma^{+}_{i}\right) \right]}{\sum_{j}\text{tr}\left[\sigma^{+}_{j}\sigma^{-}_{j}\rho_\text{ss}\right]}
$$(dimer-g1-model2)

Notice that the cross term not necessarily vanishes for $\tau \ne 0$.  Hence, there can be coherence across the emitters.   The system momentarily leaves the Gibbs state and the off-diagonal element between $|eg\rangle$ and $|ge\rangle$ grows until the system back to the steady state.

## Second-order coherence function

**Detector model I**

Using the detector model I, the correlation between two space-time points is given by

$$
G^{(2)}(\hat{\mathbf{n}}_{1},t_1;\hat{\mathbf{n}}_{2},t_2) = \sum_{ij}   \langle D^{\dagger}_{i}(\hat{\mathbf{n}}_{1},t_1)D^{\dagger}_{j}(\hat{\mathbf{n}}_{2},t_2) D_{j}(\hat{\mathbf{n}}_{2},t_2) D_{i}(\hat{\mathbf{n}}_{1},t_1)\rangle
$$(eq:G2-dimer-model1)

Notice that the summation is outside of the expectation value. We will mainly study temporal correlation measure at a single position.  By normalizing {eq}`eq:G2-dimer-model1`, we obtain the second-order coherence function 

$$
g^{(2)}(\hat{\mathbf{n}};t+\tau,t)  
= \frac{\sum_{ij} \langle\sigma^{+}_{i}(t) \sigma^{+}_{j}(t+\tau)\sigma^{-}_{j}(t+\tau)\sigma^{-}_{i}(t)\rangle_{ss}}{ \left( \sum_{i} \langle\sigma^{+}_{i}(t)\sigma^{-}_{i}(t)\rangle_{ss}\right)^2} = \frac{\sum_{ij} \text{tr}\left[\sigma^{+}_{j}\sigma^{-}_{j} e^{\mathcal{L}\tau} \left(\sigma^{-}_{i} \rho_\text{ss} \sigma^{+}_{i}\right)\right] }{ \left( \sum_{i} \langle\sigma^{+}_{i}(t)\sigma^{-}_{i}(t)\rangle_{ss}\right)^2}
$$(eq:g2-dimer-model1)



The no-delay coherence can be calculated (See {numref}`ssec:appendix-dimer-g2`) as

$$
g^{(2)}(0) = \frac{1}{2},
$$(eq:g2_0-dimer-model1)

indicating antibunching.

**Detector model II**

With the detector model II, the correlation function is

$$
G^{(2)}(\hat{\mathbf{n}}_{1},t_1;\hat{\mathbf{n}}_{2},t_2) = \langle D^{\dagger}_{i}(\hat{\mathbf{n}}_{1},t_1)D^{\dagger}(\hat{\mathbf{n}}_{2},t_2) D(\hat{\mathbf{n}}_{2},t_2) D(\hat{\mathbf{n}}_{1},t_1)\rangle
$$(eq:G2-dimer-model2)

Notice that summation over the emitters is inside $D$ and thus inside the expectation value, which allows interference.  The normalized temporal correlation is

$$
\begin{aligned}
g^{(2)}(\tau) &= \frac{ \left\langle \left[\sum_{k} e^{i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{k}}  \sigma^{+}_{k}(t)\right] \left[\sum_{\ell} e^{i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{\ell}}  \sigma^{+}_{\ell}(t+\tau)\right] \left[\sum_{i} e^{-i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{i}}  \sigma^{-}_{i}(t+\tau)\right] \left[\sum_{j} e^{-i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{j}}  \sigma^{-}_{j}(t)\right] \right\rangle_\text{ss}} 
{ \left( \sum_{i} \langle\sigma^{+}_{i}(t)\sigma^{-}_{i}(t)\rangle_{ss}\right)^2} \\
&= \frac{\sum_{ijk\ell} e^{-i k_{0} \hat{\mathbf{n}} \cdot \left(\mathbf{r}_{i}+\mathbf{r}_{j}-\mathbf{r}_{k}-\mathbf{r}_{\ell}\right)}
\langle \sigma^{+}_{k}(t)\sigma^{+}_{\ell}(t+\tau) \sigma^{-}_{i}(t+\tau)  \sigma^{-}_{j}(t)\rangle_\text{ss}} { \left( \sum_{i} \langle\sigma^{+}_{i}(t)\sigma^{-}_{i}(t)\rangle_{ss}\right)^2} \\
&=\frac{\sum_{ijk\ell} e^{-i k_{0} \hat{\mathbf{n}} \cdot \left(\mathbf{r}_{i}+\mathbf{r}_{j}-\mathbf{r}_{k}-\mathbf{r}_{\ell}\right)} \text{tr} 
\left[\sigma^{+}_{k}\sigma^{-}_{j} e^{\mathcal{L}\tau} \left(\sigma^{-}_{i} \rho_\text{ss} \sigma^{+}_{\ell}\right)\right] }{ \left(\text{tr} \left[\sigma^{+}_{i}(t)\sigma^{-}_{i}(t)\rho_{ss}\right]\right)^2}
\end{aligned}
$$(eq:g2-dimer-model2)

which has more nested summations than Eq. {eq}`eq:g2-dimer-model1`.  The no-delay coherence can be evaluated analytically (See {numref}`ssec:appendix-dimer-g2`):

$$
g^{(2)}(0) = 1,
$$(eq:g2_0-dimer-model2)

which is twice as big as Eq. {eq}`eq:g2_0-dimer-model1`.   The antibunching vanished.   

The second-order coherence function measures the interference between two photons arrived at the detector nearly at the same time.  When the detector recorded two clicks, the model I detector knows source emitter of each click and thus there is no ambiguity.   On the other hand, the detector II does not know which click corresponds to the photon from the emitter 1.  Hence, there are two possibility. The first click is from emitter 1 and the second is from emitter 2,  or vice versa.  This indistinguishability increases the chance that two photon arrive at the same time.  This is essentially the Hanbury Brown and Twiss (HBT) effect.[^1] 

[^1]: The Hanbury Brown and Twiss (HBT) effect was originally explained by classical optics. Fano{cite:p}`Fano1961` successfully explained it using the principle of quantum mechanics.

## Appendix: Calculation of $g^{(2)}(0)$

We use the following properties of the rasing and lowering operators:

1. $\sigma^{-}_{j}\sigma^{-}_{j} = 0$ and $\sigma^{+}_{j}\sigma^{+}_{j}=0$
2. $\sigma^{+}_{i}\sigma^{-}_{j} = \sigma^{-}_{j}\sigma^{+}_{i}$ for $i \ne j$.

Since the two emitters are independent, there is no cross correlation between the two emitters. The followings are useful.  These relations can be shown easily with $\rho_{ss} = \rho_{1} \otimes \rho_{2}$. 


1. $\langle \sigma^{+}_{i} \sigma^{-}_{j}\rangle_\text{ss} = 0 \quad$ for $i \ne j$.
2. $\langle \sigma^{+}_{i} \sigma^{+}_{j} \sigma^{-}_{j} \sigma^{-}_{i}\rangle_\text{ss} = \langle \sigma^{+}_{i} \sigma^{-}_{i}\rangle_\text{ss} \langle \sigma^{+}_{j} \sigma^{-}_{j}\rangle_\text{ss} \quad$ for $i \ne j$.  


In addition we have the following permutation symmetry
1. $\langle \sigma^{+}_{1} \sigma^{-}_{1}\rangle_\text{ss} = \langle \sigma^{+}_{2} \sigma^{-}_{2}\rangle_\text{ss}$
2. $\langle\sigma^{+}_{1} \sigma^{+}_{2}\sigma^{-}_{2}\sigma^{-}_{1}\rangle_\text{ss}=\langle\sigma^{+}_{2} \sigma^{+}_{1}\sigma^{-}_{1}\sigma^{-}_{2}\rangle_\text{ss}$

**Detector model I**

In the definition of the coherence function, we note that the cross term vanishes 

$$
D_{j} D_{j} = e^{-2 i k_{0} \hat{\mathbf{n}} \cdot \mathbf{r}_{j}} \sigma^{-}_{j}(t)\sigma^{-}_{j}(t) = 0 
$$

and for $i \ne j$,

$$
D_{1} D_{2} = D_{2} D_{1} = e^{-i k_{0} \hat{\mathbf{n}} \cdot ( \mathbf{r}_{1} + \mathbf{r}_{2}}) \sigma^{-}_{1}(t)\sigma^{-}_{2}(t)
$$

where $\delta_{ij}$ is the Kronecker's delta.

Hence  we have

$$
g^{(2)}(0) = \frac{2 \langle\sigma^{+}_{1} \sigma^{+}_{2}\sigma^{-}_{2}\sigma^{-}_{1}\rangle_\text{ss}}{\left( 2 \langle \sigma^{+}_{1}\sigma^{-}_{1} \rangle_\text{ss} \right)^2}
= \frac{2 \langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_\text{ss} \langle\sigma^{+}_{2}\sigma^{-}_{2}\rangle_\text{ss}}{4 \langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_\text{ss}^2} = \frac{1}{2}
$$

**Detector model II**

Similarly we note that

$$
D D = \sum_{i\ne j} e^{-i k_{0} \hat{\mathbf{n}} \cdot ( \mathbf{r}_{i} + \mathbf{r}_{j}}) \sigma^{-}_{i}(t) \sigma^{-}_{j}(t) = 2 e^{-i k_{0} \hat{\mathbf{n}} \cdot ( \mathbf{r}_{1} + \mathbf{r}_{2}}) \sigma^{-}_{1}(t) \sigma^{-}_{2}(t)
$$


The main difference between the detector models is that in the model I the summation is outside of $\langle \cdots \rangle but in the model II it is inside.

$$
g^{(2)}(0) = \frac{ 4  \langle \sigma^{+}_{1}\sigma^{+}_{2} \sigma^{-}_{2}\sigma^{-}_{1}\rangle_\text{ss}}{ \left( 2 \langle \sigma^{+}_{1}\sigma^{-}_{1} \rangle_\text{ss}\right)^2} = \frac{4 \langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_\text{ss} \langle\sigma^{+}_{2}\sigma^{-}_{2}\rangle_\text{ss}}{4 \langle\sigma^{+}_{1}\sigma^{-}_{1}\rangle_\text{ss}^2} = 1
$$
