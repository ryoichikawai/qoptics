# Dimer: The model

We investigates the coherency properties of radiation emitted by a pair (dimer) of emitters. A theoretical method for a dimer was already developed in 1970 by Lehmberg{cite:p}`Lehmberg1970b` and a comprehensive review is given in Ref. {cite:p}`Ficek2002`.    The key parameter that controls the coherency is the distance between the two emitters relative to the wave length of light they emit.  The vector from emitter 1 to emitter 2 is denoted as $\mathbf{r}_{12} = \mathbf{x}_{2} - \mathbf{x}_{1}$ and its magnitude is $r_{12}$. The wave length is denoted as $\lambda_{0} = 2\pi/k_{0} = 2 \pi c/\omega_{0}$.  Then, the key parameter is $r_{12}/\lambda_{0} = \frac{1}{2\pi} k_{0} r_{12}$. See {numref}`ssec:r12` for the typical parameter values for quantum dots. We will quantitatively investigate the coherence with respect to the parameter.

The possibility of interference in spontaneously emitter light by a dimer was discussed by Richter {cite:p}`Richter1991` and it was observed by Eichmann _et al_.{cite:p}`Eichmann1993`. The result was theoretically analyzed by Wong _et al_.{cite:p}`Wong1997`.    Skornia _et al_.{cite:p}`Skornia2001` theoretically showed that even more interesting interference patterns can be seen when the dimer is illuminated by a laser beam.   Recently, Wolf _et al_.{cite:p}`Wolf2020` and Richter _et al_.{cite:p}`Richter2023` experimentally observed it.  We shall study the details of the interference in this note.

We compute the reduced density of the emitters using the quantum master equation introduced in {numref}`ch:openqm-qme`. The coherence functions are evaluated based using the far field modeled by Glauber's detection theory.  Furthermore, we unravel the dynamics using the jump method introduced in {numref}`ch:openqm-jump`.  A summary of method will be given in next section.

We discuss three cases in details.

1. **Thermal fluorescence from an incoherent pair**  ($r_{12}/\lambda_{0} \gg 1$)
2. **Thermal fluorescence from a coherent pair: Superradiance** ($r_{12}/\lambda_{0} \ll 1$)
3. **Coherence by driving: Interference and bunching/antibunching** ($r_{12}/\lambda_{0} \gg 1$)
