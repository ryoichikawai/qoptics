# Dimer: The model

We investigates the properties of radiation emitted by a pair (dimer) of emitters. A theoretical method for a dimer was already developed in 1970 by Lehmberg{cite:p}`Lehmberg1970b` and a comprehensive review is given in Ref. {cite:p}`Ficek2002`.    The key parameter is the distance between the two emitters relative to the wave length of light they emit, $r_{12}/\lambda_{0} = \frac{1}{2\pi} k_{0} r_{12}$. See {numref}`ssec:r12` for the typical parameter values for quantum dots. We will quantitatively investigate the coherence with respect to the parameter.

The possibility of interference in spontaneously emitter light by a dimer was discussed by Richter {cite:p}`Richter1991` and it was observed by Eichmann _et al_.{cite:p}`Eichmann1993`. The result was theoretically analyzed by Wong _et al_.{cite:p}`Wong1997`.    Skornia _et al_.{cite:p}`Skornia2001` theoretically showed that even more interesting interference patterns can be seen when the dimer is illuminated by a laser beam.   Recently, Wolf _et al_.{cite:p}`Wolf2020` and Richter _et al_.{cite:p}`Richter2023` experimentally observed it.  We shall study the details of the interference in this note.

We compute the reduced density of the emitters using the quantum master equation introduced in {numref}`ch:openqm-qme`. The coherence functions are evaluated based using the far field modeled by Glauber's detection theory.  Furthermore, we unravel the dynamics using the jump method introduced in {numref}`ch:openqm-jump`.  A summary of method will be given in next section.

## Symmetry

A pair of atoms have a very high symmetry.  Without the dipole moment, its point group is  $D_{\infty h}$.  The symmetry is reduced when the direction of the dipole moment is takeon into account as {numref}`tbl:dimer-symmetry` shows.   The point group contains group elements which leave both atoms exactly where they are,  for example, the rotation about the molecular axis ($c_{\infty}$). We really don't need them.  What we need is permutation symmetry.  In all cases, it is $S_{2}$ since all emitters carry the same dipole moment.  This simplifies theoretical investigation very simple.  For example, the Decke basis for $N=2$ works in any direction of the dipole moment.  Even when the dipole-dipole moment is taken into account, the Decke basis works.  Only when the emitters is under external  force, we need to use different basis a set.


We discuss three cases in details.

1. **Thermal fluorescence from an incoherent pair**  ($r_{12}/\lambda_{0} \gg 1$)
2. **Thermal fluorescence from an incoherent pair with dipole-dipole coupling:** ($r_{12}/\lambda_{0} \ll 1$)
3. **Coherence by driving: Interference and bunching/antibunching** ($r_{12}/\lambda_{0} \gg 1$)

## Model I: Thermal fluorescence without the dipole-dipole interaction

When the distance between the emitters is much larger than the wave length.  Then, the dipole-dipole coupling is negligible. More precisely, we assume $\Omega < \omega_{0}$ and also $\Omega < T$ where $\Omega$ is the coupling strength. In this case, either the product basis or the Dicke basis give the same results.  The steady state is a thermal state, which is diagonal in either basis set.  Hence, there is no coherency in them.  When the separation is large, the detector may be able to distinguish the radiation from individual emitters.  If so, each beam behave exactly in the same way as a single emitter.  However, we can combine the signal after the measurement.  On the other hand, the detector can measure photons in all beams without distinguishing their sources.  Then, we have different statistics.

## Model II:  Thermal fluorescence with the dipole-dipole interaction

When the distance between the emitters is shorter than the wave length, the dipole-dipole interaction is no longer negligible. It shifts energy levels shift and the product basis vectors are no longer the energy eigenstates. We must use the Dicke basis.  The transition energy is no longer $\omega_{0}$.  Instead we have two different transition energies $\omega_{0} \pm \Omega$.  The simple quantum master equation does not take into account  this energy shift.   We first use the traditional quantum master equation ignoring the energy shift.  In other words, the decay rate changes only due to the symmetry consideration. This approxaimation breaks the detailed balance and thus the steady state is not the Gibbs state.  Nevertheless, we expect that this approximation is useful for the qualitative understanding of super/subradiance processes, which is our main goal.   What we miss is the coherence developed between $|ge\rangle$ and $|eg\rangle$ by the coupling.  In other words, the steady state is diagonal in the Dicke basis but not in the product basis.   Hence, the result of the simple quantum master equation is quantitatively incorrect.  We need to use a slightly higher level of model known as the GKSL master equation{cite:p}`Gorini1976,Lindblad1976`, in particular the Davies's weak-coupling GKSL master equation{cite:p}`Davies1974`.[^1]   We will compare the results from the standard QME and GKSL QME.  Another issue arises when the Dicke limit is taken.  Since the subradiant channel becomes totally dark.  The ergodicity is broekn and the system will never be thermalized.  Even when the ergodicity is not exactly broken, the decay rate of the subradiant channel becomes so small that it takes long time to reach a steady state (thermal state). 

[^1]: The microscopic derivation of a quantum master equation given in Ref. {cite:p}`Breuer2002` is essentially the Davies'method.

## Model III: Coherence induced by external driving

When the two emitters are driven by the same laser beam, they are "synchronized" and establish coherence in between product states $|ge\rangle$ and $|eg\rangle$.  This coherence generates interesting interference patterns, which is originally predicted by {cite:p}`Skornia2001` and experimentally observed by {cite:p}`Wolf2020`.  Their resutls are recreated numerically.  Neither product basis nor standar Dicke basis works for this system.  However, a Dicke-like basis set, which we shall call Skornia basis, exhibits super/subradiant channels, which is investigated in {cite:p}`Richter2023`.
