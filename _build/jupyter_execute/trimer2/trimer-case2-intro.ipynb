# Trimer Case II:  Full S₃ symmetry

We continue to investigate thermal fluorescence of a trimer.  In this chapter, the emitters form an equilateral triangle
$r_{12} = r_{23} = r_{31}$.  Hence, geometrically the structure belongs to point group $D_{3h}$ symmetry group.  However, the symmetry of the whole system including the radiation field can be quite different depending on the orientation of the dipole moment.  We also need to recall that the orientation of the dipole moment is identical to all emitters.

When the dipole moment is perpendicular to the plane of the triangle (See {numref}`fig:equilateral-tirangle-D_3`), the space inversion with respect to $z=0$ reverse the direction of the dipole moment.  Hence its symmetry is $D_{3}$ which is isomorphic to $S_3$.

```{figure}  equilateral-triangle-D_3.webp
:alt: An equilateral triangle with dipole moment perpendicular to the plane
:align: center
:name: fig:equilateral-tirangle-D_3.2

When the dipole moment is perpendicular to the plane, the inversion symmetry with respect to $z=0$ is lost.  Hence, the symmetry group is $D_3$ rather than $D_{3h}$.
```

Its high symmetry makes theory rather simple. We can predict various interesting optical properties, which we discuss in the next section.  The theoretical preictions are numerically confirmed using quantum master equation.

There not many literature which investigate optical properties of $S_3$ system. See for example Ref. {cite:p}`Feng2013`.
