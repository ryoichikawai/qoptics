# Optical quantum master equations

## Full Hamiltonian

In order to investigate the optical properties of emitters, we need a Hamiltonian consisting of three parts, $H_{S}$ for the system of emitters, $H_{R}$ for the radiation field, and $H_{SR} for the coupling between them.  They are already introduced in {numref}`sec:emitter`, {numref}`sec:radiation-field`}, and {numref}`sec:E1-coupling`.

Formally, we are supposed to solve the Liouville-von Neumann equation $\rho_{SR} = -i \left[H_{S}+H_{R}+H_{SR},\rho_{SR}\right]$ but it is mathematically quite demanding.  Hence, we need some approximations.   

There are two effects of the interaction with the radiation field on the emitters.  One is the energy shift including the Lamb shift and the dipole-dipole coupling, which is the real part of the self energy.    The other is the dissipation which arises from the imaginary part of the self energy.  The result of the approximation is given below.

## Quantum master equation: A general form

If the interaction between the system and the radiation field is weak, we can use the perturbation theory (the Born approximation).  We also assume the radiation field goes back to the thermal equilibrium instantly after a photon is emitted. (Markovian approximation).  Furthermore, we adopt the rotating wave approximation  With these approximation, we can calculate the evolution of the emitters and the radiation field separately. For the emitters, the evolution of the reduced density $\rho$ for the emitters is determined by the quantum master equation, which is also known as the Lindblad equation $\frac{d}{dt}\rho = \mathcal{L} (\rho)$ where $\mathcal{L}$ is the Liouvillian superoperator which consists of two parts, a unitary part (also called coherent part) and dissipator:

$$
\mathcal{L} (\rho) = -i [H_{S}+H^\prime_{S}, \rho] + \mathcal{D} (\rho)
$$(eq:qme-general)

The first term in the right hand side is the unitary part, which includes additional Hamiltonian $H^\prime_{S}$.   The second term is the dissipator expressed as a super operator which takes a general form:

$$
\mathcal{D} (\rho) = \sum_{ij} \Gamma_{ij} \left[L_{i} \rho L_{j}^\dagger - \frac{1}{2} \left\{L_{j}^\dagger L_{i}, \rho\right\} \right]
$$(eq:dissipator-general)

where $\{A,B\} = AB + BA$ is the anticommutator.  The operator $L_{j}$ is known as Lindblad operator or jump operator[^1]  and $\Gamma_{ij}$ is the transition rate.

We can always diagonalize $\Gamma_{ij}$ and make the disspator in a diagonal form:

$$
\mathcal{D} (\rho) = \sum_{j} \Gamma_{j} \left[\tilde{L}_{j} \rho \tilde{L}_{j}^\dagger - \frac{1}{2} \left\{\tilde{L}_{j}^\dagger \tilde{L}_{j}, \rho\right\} \right]
$$(eq:dissipator-diagonal)

For small systems, group theory helps to diagonalize the matrix but for larger ones, we may need numerical diagonalization.

[^1]:  In some literature, it is also called _collapse operator_.  However, we reserve this name for $C_j = \sqrt{\Gamma_{j}} L_{j}$.

## Ambiguity

The expression of a quantum master equation is not unique.  It is invariant under many transformation.  Suppose that the dissipator is given by

$$
\mathcal{D}(\rho) = \sum_{ij} \Gamma^{A}_{ij} \left(A_{j} \rho A^{\dagger}_{i} - \frac{1}{2} \left \{A^{\dagger}_{i}A_{j},\rho\right\}\right)
$$

Introducing a set of new operators using a unitary transformation $U$ such that $A_{i} = \sum_{k} U_{ik} B_{k}$, the dissipator is transformed to

$$
\begin{aligned}
\mathcal{D}(\rho) &= \sum_{ij} \Gamma^{A}_{ij} \left[ \left(\sum_{k} U_{jk} B_{k}\right) \rho \left(\sum_{\ell} U^{*}_{i\ell} B^{\dagger}_{\ell}\right) - \frac{1}{2} \left \{\left(\sum_{\ell} U^{*}_{i\ell} B^{\dagger}_{\ell}\right) \left(\sum_{k} U_{jk} B_{k}\right),\rho\right\}\right] \\
&= \sum_{k\ell} \Gamma^{B}_{k\ell} \left(B_{k} \rho B^{\dagger}_{\ell} - \frac{1}{2} \left \{B^{\dagger}_{\ell}B_{k},\rho\right\}\right)
\end{aligned}
$$

where

$$
\Gamma^{B}_{k\ell} = \sum_{ij} U_{k,i}  \Gamma^{A}_{ij} U^{*}_{i,\ell}.
$$


Even under a simple shift $A_{j} = B_{j} + \alpha_{j}$ where $\alpha_{j}$ is a constant, the dissipator is invariant.  Hence, any isomtric transformation preserves the dissipator and there are infinitely many expression of the same quantum master equation.  



## Diagonalization

Utilizing the above freedom, we can always digaonalize the dissipator.

Let $\lambda_{i}$ and $\mathbf{v}_{i}$ be eigenvalues and corresponding eigenvectors of $\Gamma_{ij}$.   The diagonalization is done by the unitary transformation $U = [\mathbf{v}_{1}, \mathbf{v}_{2}, \cdots, \mathbf{v}_{N}]$. The new transition rate matrix is $\Gamma^{B} = U^{\dagger} \Gamma^{A} U = \text{diag}(\lambda_1, \lambda_2, \cdots, \lambda_N)$.  At the same time, the operator is transformed as $B_{k} = \sum_{j} U^{\dagger}_{kj} A_{j} = \sum_{j}  v^{k}_{j} A_{j}$ where $v^{k}_{j}$ is the $k$-th component of the $j$-th eigenvector $\mathbf{v}_{j}$.    

Since the transformation does not change the dissipator, either expression generates the same solution.  The diagonal form has a slight computational advantage[^].


As an example, consider a pair of two-level emitters.  The dissipator is given by:

$$
\mathcal{D} \rho = \sum_{ij} \Gamma_{ij}  \left( \sigma^{+}_{j} \rho \sigma^{-}_{i} - \frac{1}{2} \left\{\sigma^{-}_{i}\sigma^{+}_{j}, \rho\right\} \right).
$$

The matrix of transition rates is

$$
\Gamma = \begin{pmatrix} \Gamma_{11} & \Gamma_{12} \\ \Gamma_{21} & \Gamma_{22} \end{pmatrix} =  \begin{pmatrix} a & b \\ b & a \end{pmatrix}
$$

where we used the symmetric property (permutation symmetry) $a \equiv \Gamma_{11} = \Gamma{22}$ and $b=\Gamma_{12}=\Gamma_{21}$.  The eigenvalues are $\lambda_{1} = a+b$ and $\lambda_{2} = a-b$ and the corresponding eigenvectors are $\mathbf{v}_{1} = \tfrac{1}{\sqrt{2}} \begin{pmatrix}1 \\ 1\end{pmatrix}$ and $\mathbf{v}_{2} = \tfrac{1}{\sqrt{2}} \begin{pmatrix}1 \\ -1\end{pmatrix}$.  USeing the eivenvectors, we  obtain new operators

$$
\eta^{\pm}_{1} = (\sigma^{\pm}_{1} + \sigma^{\pm}_{2})/\sqrt{2}, \qquad \eta^{\pm}_{2} = (\sigma^{\pm}_{1} - \sigma^{\pm}_{2})/\sqrt{2}
$$(eq:dissipator-sigma)

The original dissipator is transformed to 

$$
\mathcal{D} (\rho) = (\Gamma_{11}+\Gamma_{12}) \left( \eta^{+}_{1} \rho \eta^{-}_{1} - \frac{1}{2} \left\{\eta^{-}_{1}\eta^{+}_{1}, \rho\right\} \right) + (\Gamma_{11}-\Gamma_{12}) \left( \eta^{+}_{2} \rho \eta^{-}_{2} - \frac{1}{2} \left\{\eta^{-}_{2}\eta^{+}_{2}, \rho\right\} \right).
$$(eq:dissipator-eta)

Notice that there is no cross terms between $\eta^{\pm}_{1}$ and $\eta^{\mp}_{2}$.  


This transformation does not change the Hamiltonian and we have not changed the basis set for the Hilbert space.  However, since we took into account the permutation symmetry to get the diagonal form of the dissipator, it is nicer to use a symmetry adopted basis set.  In the current example, the Hamiltonian is given by

$$
H = \frac{\omega_{0}}{2} \left(\sigma^{z}_{1} + \sigma^{z}_{2} \right) + \Omega \left(\sigma^{+}_{1}\sigma^{-}_{2} + \sigma^{+}_{2}\sigma^{-}_{1}\right)
$$

which is invariant under permutation symmetry $S_{2}$.  Hence, the symmetry adopted basis set $|e\rangle = |ee\rangle$, $|s\rangle = \frac{1}{\sqrt{2}}\left(|eg\rangle+|ge\rangle\right)$, $|a\rangle = \frac{1}{\sqrt{2}}\left(|eg\rangle-|ge\rangle\right)$, and $|g\rangle = |gg\rangle$ are the energy eigenvectors of the system.  

While  $\Gamma$ is $2\times2$, the Hamiltonian is $4 \times 4$. The number of eigenvectors are different. However, since both the Hamiltonian and the dissipator belong to the same symmetry group, we can diagonalize both using the group theory.  We will discuss the details in a later section.


[^1]: Especially when `mesolve` in `QuTiP` is used since many functions assume the diagonal form.

## Optical master equation


As introduced in {numref}`sec:emitter`, {numref}`sec:radiation-field`, and {numref}`sec:E1-coupling`, the Hamiltonian of the whole system is given by $H=H_{0}+H_{R}+H_{SR}$, we obtain the quantum master equation for the emitters coupled to the radiation field by tracing out the degrees of freedom of radiation fields.  

First, two terms are added to the free Hamiltonian $H_{0}$.  The net Hamiltonian is  

$$
H_{S} = H_{0} + H_\text{LS} + V_{dd}
$$

with the Lamb shift $H_\text{LS}$

$$
H_\text{LS} = \sum_{j} \Omega_{jj} \sigma^{+}_{j}\sigma^{-}_{j}
$$(eq:H_LS)

and the dipole-dipole interaction
 
$$
V_\text{dd} = \sum_{i \ne j} \Omega_{ij} \sigma^{+}_{i}\sigma^{-}_{j}.
$$(eq:H_dd)

The matrix $\Omega_{ij}$ is defined in the next subsections.  The Lamb shift ($\Omega_{jj}$) diverges in the non-relativistic quantum mechanics. The renormalization of electron mass and charge in QFT resolves this issue. It turns out that the shift is very small and thus we shall ignore it.

Secondly, we obtain a dissipator:

$$
\begin{aligned}
\mathcal{D}(\rho) &= \sum_{i,j} \Gamma_{ij} \left[N(\omega_{0})+1\right] \left[\sigma^{-}_{j} \rho \sigma^{+}_{i} - \frac{1}{2}\left\{\sigma^{+}_{i}\sigma^{-}_{j}, \rho\right\} \right] \\
&+ \sum_{i,j}\Gamma_{ij} N(\omega_{0}) \left[\sigma^{+}_{j} \rho \sigma^{-}_{i} - \frac{1}{2}\left\{\sigma^{-}_{i}\sigma^{+}_{j}, \rho\right\} \right]
\end{aligned}
$$(eq:optical-dissipator)

where $N(\omega) = \frac{1}{e^{\omega/T}-1}$ is the mean thermal photon number at temperature $T$.   
The first term in the right hand side corresponds to the emission and the second term to the absorption.  The explicit definition of $\Gamma_{ij}$ is given later in this page.

Combining all new terms, we find an optical master equation

$$
\frac{d}{dt} \rho = -i [H_{S} + H_\text{LS} + V_{dd},\rho] + \mathcal{D} (\rho)
$$(eq:optical-qme)

We will be using this master equation through out this lecture.

## Dipole-diploe coupling

The coefficient of the H$_{dd}$ {eq}`eq:H_dd` is given by

$$
\Omega_{i\ne j} = - \frac{3\gamma_0}{4} \left[ \left(1 - \cos^2 \theta_{ij}\right) \frac{\cos(k_0 r_{ij})}{k_0 r_{ij}}
+ \left(1-3 \cos^2 \theta_{ij} \right) \left (\frac{\sin(k_0 r_{ij})}{(k_0 r_{ij})^2} -  \frac{\cos(k_0 r_{ij})}{(k_0 r_{ij})^3} \right) \right]
$$(eq:omega_ij)

where $\theta_{ij}$ is the angle betwen $\mathbf{d}$ and $\mathbf{r}_{ij}$ and the rate of spontaneous emission at zero temperature is defined by  

$$
\gamma^{0} = \frac{4 \omega_{0}^{3} |\vec{d}|^2}{3\hbar c^3}.
$$ 

$\Omega_{ij}$ diverges as $r_{ij} \rightarrow 0$ by the same as the divergence of the Lamb shift.   We cannot the dipole-dipole interaction when the two emitters are too close.

The main role of this coupling is to shift the energy eigenvalue of the emitters.  For a pair of emitters, the spectra of $H_{S}$ is
$E_{ee}=\omega_0$, $E_{eg}=E_{ge}=0$, and  $E_{gg}=-\omega_0$ in absence of $H_{dd}$. The corresponding energy eigenvectors are just the product basis vectors.  When $H_{dd}$ is added, the degeneracy is lifted and the new energy eigenvalues are $E_{e} = \omega_{0}$, $E_{s} = \Omega_{12}$, $E_{a} = - \Omega_{12}$ and $E_{g} = -\omega_{0}$.  The corresponding eigenstates are $|e\rangle = |ee\rangle$, $|s\rangle = (|eg\rangle + |ge\rangle)/\sqrt{2}$, $|a\rangle = (|eg\rangle - |ge\rangle)/\sqrt{2}$ and $|g\rangle = |ee\rangle$.  Hence, the optical transitions take place between these energy eigenstates.  The energy shift also creates coherence between $|eg\rangle$ and |ge\rangle$, which plays important roles.


The following code plots the strength of the dipole-dipole coupling with respect to the transition energy $\omega_0$.  The method is valid only when $\Omega \ll \omega_0$. Hence,  $r_{12}/\lambda_{0} < 0.1$ must be avoided. 

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x,theta):
    y=(1-np.cos(theta)**2)*np.cos(x)/x + (1-3*np.sin(theta)**2)*(np.sin(x)/x**2 - np.cos(x)/x**3)
    return -y*3/4
   
x0=np.linspace(0.1,2.0,100)  # distance as r/lambda_0
x1 = x0*(2*np.pi)   # distance as k_0*r
y1 = f(x1,0)
y2 = f(x1,np.pi/4)
y3 = f(x1,np.pi/2)
plt.figure(figsize=(4,3))
plt.title("strength of dipope-dipole coupling")
plt.plot(x0,y1,label=r"$\theta=0$")
plt.plot(x0,y2,label=r"$\theta=\pi/4$")
plt.plot(x0,y3,label=r"$\theta=\pi/2$")
#plt.ylim([-30,30])
plt.xlabel(r"$r_{12}/\lambda_{0}$")
plt.ylabel(r"$\Omega_{12}/\gamma_{0}$")
plt.axhline(y=0, color='grey', linestyle='--')
#plt.axvline(x=0.1, color='grey', linestyle='-.')
plt.legend(loc=4)
plt.show()
```

There is a big problem with the above formula. As the separation of two emitters vanishes, $\Omega_{12}$ diverges.   For $r_{12} > 0.01 \lambda_0$, the error is known to be less than 10\%.  When the distance is so small, two emitters nearly in contact and thus we cannot identify each emitter any longer.  Therefore, we don't worry about such short distance in realistic application.  For some theoretical interest, we may take so-called the Decke limit $r_{ij} \rightarrow 0$.  In that case, we just use a finite value.

## Decay rates

The diagonal coefficients in the dissipator {eq}`eq:optical-dissipator` are simply $\Gamma_{ii} = \gamma_{0}$ and the off-diagonal coefficients are defined by

$$
\Gamma_{i\ne j} & =  \frac{3\gamma_0}{2} \left[ \left(1 - \cos^2 \theta_{ij}\right) \frac{\cos(k_0 r_{ij})}{k_0 r_{ij}}
+ \left(1-3 \cos^2 \theta_{ij} \right) \left (\frac{\sin(k_0 r_{ij})}{(k_0 r_{ij})^2} -  \frac{\cos(k_0 r_{ij})}{(k_0 r_{ij})^2} \right) \right]
$$(eq:Gamma_ij)

As the following plot indicates, $\Gamma_{12}$ vanishes when $r_{ij}/\lambda_{0} \gg 1$ and approaches $\gamma_0$ when $r_{ij}/\lambda_{0} \ll 1$,  where $\lambda_{0} = 2\pi/k_{0} = 2 \pi c/\omega_0$.


The following code plots $\Gamma_{i\ne j}$ as the function of $r_{ij}/\lambda_{0}$.  The plot clearly shows that as $r_{ij} \rightarrow 0$, $\Gamma_{i\ne j} \rightarrow \gamma_0$, which indicates that when the radiation field cannot distinguish the emitters, all the coefficients are just $\gamma_0$.  On the other hand, if $r_{ij} \gg \lambda_0$, $\Gamma_{i\ne j} \approx 0$, which indicate that the cross correlation between the emitters $i$ and $j$ vanishes.  In conclusion, the indistinguishability of emitters by the radiation field is reflected in $\Gamma_{ij}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x,theta):
    y=(1-np.cos(theta)**2)*np.sin(x)/x + (1-3*np.cos(theta)**2)*(np.cos(x)/x**2 - np.sin(x)/x**3)
    return y*3/2

x0=np.linspace(0.1,2.0,100)  # distance as r/lambda_0
x1 = x0*(2*np.pi)   # distance as k_0*r
y1 = f(x1,0)
y2 = f(x1,np.pi/4)
y3 = f(x1,np.pi/2)
plt.figure(figsize=(4,3))
plt.title(r"Off-diagonale element $\Gamma_{ij}$")
plt.plot(x0,y1,label=r"$\theta=0$")
plt.plot(x0,y2,label=r"$\theta=\pi/4$")
plt.plot(x0,y3,label=r"$\theta=\pi/2$")
plt.xlabel(r"$r_{12}/\lambda_{0}$")
plt.ylabel(r"$\Gamma_{12}/\gamma_{0}$")
plt.axhline(y=0, color='k', linestyle='--')
plt.legend(loc=1)
plt.show()
```

## Phonon scattering

Interaction with phonon may cause a significant impact on coherence for quantum dots. The quantum master equation acuires additional terms from the interaction Hamiltonian {eq}`eq:H_phi`.  First, there is a new dissipator

$$
\mathcal{D}^{\phi} (\rho) = - \sum_{ij} \Gamma^{\phi}_{ij} \left ( \sigma^{z}_{i} \rho \sigma^{z}_{j} - \frac{1}{2} \left\{\sigma^{z}_{j}\sigma^{z}_{i},\rho\right\} \right)
$$

which could play a significant role in decoherence.  We will see if this dissipation has some impact on coherency in applications.

Secondly, a new coupling Hamiltonian shows up:

$$
H_{\phi-\phi} = \sum_{i\ne j} \Gamma(\mathbf{x}_{i} - \mathbf{x}_{j}) \sigma^{z}_{i}\sigma^{z}_{j}
$$

with the coupling strength is given by

$$
\Gamma(\mathbf{x}) = \sum_{\mathbf{q}} \frac{2|g_\mathbf{q}|^2}{\omega_{\mathbf{q}}} \cos(\mathbf{q}\cdot\mathbf{x})
$$

where $\omega_\mathbf{q}$ is the energy of phonon with momentum $\mathbf{q}$.

Since this term commutes with $H_{0}$, no energy exchange is involved.  They expected to induce pure dephasing. 
However, it seems that the effect of dephasing is not significant in many applications since competing couplings such as quantum tunneling is stronger.
