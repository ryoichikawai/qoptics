# Davies-GKSL master equation

## Issues with the optical quantum master equation

The quantum master equation is based on the second order perturbation theory along with Markovian and other approximations.  It is not always perfect.  The definition of $\Omega_{ij}$ and $\Gamma_{ij}$ in Eqs. {eq}`eq:omega_ij` and {eq}`eq:Gamma_ij`, we used the free spontaneous emission rate $\gamma_0$ using a single Bohr frequency $\omega_0$.  When emitters are free (no mutual interaction nor external potential), there is no issue.  However, when the free Hamiltonian is perturbed by the coupling, the transition energy is no longer $\omega_{0}$.  There could be multiple Bohr frequencies.  Then, each transition should have different transition rates. But the master equation {eq}`eq:optical-qme` uses a single Bohr frequency $\omega_{0}$ regardless of the actual Bohr frequencies.  

**Broken detailed balance**

As a consequence,  the detailed balance is broken and the steady state does not match to the true Gibbs state.  We might miss some coherence due to errors in the steady state.  In realistic cases, the dipole-dipole interaction is very weak with respect to $\omega_{0}$.  Hence, the energy shift due to the dipole-dipole coupling is negligible and the single Bohr frequency $\omega_{0}$ is acceptable.

**Interaction picture**

When there are multiple time scales, we want to eliminate fast time scale by using a rotating frame.  However, when there are multiple Bohr frequencies, eliminating one frequency $\omega_{0}$ does not remove all fast oscillations.  


The Davies-GKSL master equation takes into account the multiple Bohr frequencies and resolves the above issues.

## Davies-GKSL master equation

It is possible to construct a quantum master equation that satisfies the detailed balance.  The basic theory was developed in {cite:p}`Davies1974,Gorini1976,Lindblad1976`, which is known as the GKSL master equation or the Davies master equation (We shall call it the GKSL master equation).  Microscopic derivation of the GKSL optical master equation is given in {cite:p}`Breuer2002`.

Consider the Bohr frequency $\omega_{\alpha\beta}=E_{\alpha}-E_{\beta}$ between energy eigenstate $|E_{\alpha}\rangle$ and $|E_{\beta}\rangle$.  Then, we define jump operators for emission

$$
A_{j}(\omega_{\alpha\beta}) = |E_{\beta}\rangle\langle E_{\beta}| \sigma^{-}_{j}|E_{\alpha}\rangle\langle E_{\alpha}|
$$

and for absorption

$$
A_{j}(\omega_{\beta\alpha}) =  |E_{\alpha}\rangle\langle E_{\alpha}| \sigma^{+}_{j}|E_{\beta}\rangle\langle E_{\beta}|.
$$

Noting that $\omega_{\alpha\beta} = - \omega_{\beta\alpha}$, we have $A^{\dagger}(\omega_{\alpha\beta})=A(\omega_{\beta\alpha}) = A(-\omega_{\alpha\beta})$.

and the dissipator is defined as

$$
\begin{aligned}
D(\rho) =& \sum_{ij} \sum_{\omega_{\alpha\beta}>0} \Gamma_{ij}^{\downarrow}(\omega_{\alpha\beta}) \left[ A_{i}(\omega_{\alpha\beta}) \rho A^{\dagger}_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A^{\dagger}_{j}(\omega_{\alpha\beta})  A_{i}(\omega_{\alpha\beta}),\rho\right\}\right] \\
& + \sum_{ij} \sum_{\omega_{\alpha\beta}>0} \Gamma_{ij}^{\uparrow}(\omega_{\alpha\beta}) \left[ A^{\dagger}_{i}(\omega_{\alpha\beta}) \rho A_{j}(\omega_{\alpha\beta}) - \frac{1}{2}\left\{A_{j}(\omega_{\alpha\beta})  A^{\dagger}_{i}(\omega_{\alpha\beta}),\rho\right\}\right] 
\end{aligned}
$$(eq:GKSL-dissipator)

where emission rate $\Gamma^{\downarrow}_{ij}$ and absorption rate $\Gamma^{\uparrow}_{ij}$ are defined by 

$$
\begin{aligned}
\Gamma^{\downarrow}_{ij} (\omega_{\alpha\beta}) &= \frac{4 d^2 \omega_{\alpha\beta}^3}{3 \hbar c} \left[1 + \bar{n}(\omega_{\alpha\beta})\right]\\
\Gamma^{\uparrow}_{ij} (\omega_{\alpha\beta}) &= \frac{4 d^2 \omega_{\alpha\beta}^3}{3 \hbar c} \bar{n}(\omega_{\alpha\beta})
\end{aligned}
$$(eq:GKSL-Gamma)

with $N(\omega) = \left(e^{\hbar \omega/T}-1\right)^{-1}$.

These transition rate satisfies the detailed balance 

$$
\frac{\Gamma^{\uparrow}(\omega_{\alpha\beta})}{\Gamma^{\downarrow}(\omega_{\alpha\beta})} = \frac{\bar{n}(\omega_{\alpha\beta})}{1 + \bar{n}(\omega_{\alpha\beta})} = e^{-\omega_{\alpha\beta}/T}
$$

which guarantees that the steady state is the Gibbs state.

In conclusion, when the dipole-dipole coupling is weak, that is $\Omega_{ij} \ll \omega_{0}$ and also $\Omega_{ij} \ll T$, we use the standard optical master equation. Otherwise, we use the GKSL version with {eq}`eq:GKSL-dissipator` and {eq}`eq:GKSL-Gamma`.

The jump operators $A_{j}(\omega)$  sattisfies $[H,A_{j}(\omega)] = - \omega A(\omega)$.   Hence, $e^{-i H t} A_{j}(\omega) e^{i H t} = e^{-i \omega} A(\omega)$.  This property is essential when the full interaction picture is used to eliminate rapid oscillation. See {numref}`sec:openqm-qme-rotating-frame`.

## The number of jump operators

In general, there is a jump operator for each transition.  That means For N=2, there for 4 jump operators for emission and another 4 operators for absorption, totaling 8 operators.  The standard optical master equation had only 4.  For N=3 with $S_3$ symmetry,  we need 14 operators and if there is no symmetry there are 26 operators. The optical master equation needs only 6.   The number of jump operators grows very rapidly as the number of emitters increases.  Hence, this method is not practical for large $N$ unless the system has a high level of symmetry.
