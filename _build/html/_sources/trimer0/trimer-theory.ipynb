# Trimer:  Theoretical model

<span style="color:red">**This page is still under construction.**</span>




We consider a case where the distance between every pair of emitters is identical, that is $r_{12} = r_{23} = r_{31}$.  Hence, geometrically the structure belongs to $D_{3h}$ symmetry group.  However, the symmetry of the whole system including the radiation field can be quite different depending on the orientation of the dipole moment.  We also need to recall that the orientation of the dipole moment is identical to all emitters.

When the dipole moment is perpendicular to the plane of the triangle (See {numref}`fig:equilateral-tirangle-D_3`), the space inversion with respect to $z=0$ reverse the direction of the dipole moment.  Hence its symmetry is $D_{3}$ which is isomorphic to $S_3$.

```{figure}  equilateral-triangle-D_3.webp
:width: 3in
:alt: An equilateral triangle with dipole moment perpendicular to the plane
:align: center
:name: fig:equilateral-tirangle-D_3

When the dipole moment is perpendicular to the plane, the inversion symmetry with respect to $z=0$ is lost.  Hence, the symmetry group is $D_3$ rather than $D_{3h}$.
```

If the dipole moment is in-plane, it is not possible to have a high symmetry such as $S_{3}$, $D_{3}$ nor $C_{3}$. When the moment is parallel to a symmetric axis (the left triangle in {numref}`fig:equilateral-tirangle-C_2`),  mirror symmetry with respect to $y$ axis is preserved.  Hence, it has $C_{2}$ symmetry.  Otherwise, there is no symmetry and it belong to $C_{1}$ group.  Hence, it is similar to a scalene triangle.

```{figure}  equilateral-triangle-C_2.webp
:width: 6in
:alt: An equilateral triangle with dipole moment in the plane
:align: center
:name: fig:equilateral-tirangle-C_2

_Left_: the dipole moment is in-plane and parallel to a symmetric axis. Its symmetry is equivalent to an isosceles triangle ($C_2$).  _Right_:   The dipole moment is not parallel to any symmetric axis.  There is no symmtry and the situation is similar to a scalene triangle.
```
