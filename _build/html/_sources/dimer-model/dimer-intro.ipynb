# Dimer: The model

We investigates the properties of radiation emitted by a pair (dimer) of emitters. A theoretical method for a dimer was already developed in 1970 by Lehmberg{cite:p}`Lehmberg1970b` and a comprehensive review is given in Ref. {cite:p}`Ficek2002`.    The key parameter is the distance between the two emitters relative to the wave length of light they emit, $r_{12}/\lambda_{0}$. See {numref}`sec:openqm-qme-scaling` for the typical parameter values for quantum dots. We will quantitatively investigate the coherence with respect to the parameter.

The possibility of interference in spontaneously emitter light by a dimer was discussed by Richter {cite:p}`Richter1991` and it was observed by Eichmann _et al_.{cite:p}`Eichmann1993`. The result was theoretically analyzed by Wong _et al_.{cite:p}`Wong1997`.    Skornia _et al_.{cite:p}`Skornia2001` theoretically showed that even more interesting interference patterns can be seen when the dimer is illuminated by a laser beam.   Recently, Wolf _et al_.{cite:p}`Wolf2020` and Richter _et al_.{cite:p}`Richter2023` experimentally observed it.  We shall study the details of the interference in this note.

We compute the reduced density of the emitters using the quantum master equation introduced in {numref}`ch:openqm-qme`. The coherence functions are evaluated based using the far field modeled by Glauber's detection theory.  Furthermore, we unravel the dynamics using the jump method introduced in {numref}`ch:openqm-jump`.  A summary of method will be given in next section.

## Symmetry

A pair of atoms have a very high symmetry.  Without the dipole moment, its point group is  $D_{\infty h}$.  The symmetry is reduced when the direction of the dipole moment is takeon into account as {numref}`tbl:dimer-symmetry` shows.   The point group contains group elements which leave both atoms exactly where they are,  for example, the rotation about the molecular axis ($c_{\infty}$). We really don't need them.  What we need is permutation symmetry.  In all cases, it is $S_{2}$ since all emitters carry the same dipole moment.  This simplifies theoretical investigation very simple.  For example, the Decke basis for $N=2$ works in any direction of the dipole moment.  Even when the dipole-dipole moment is taken into account, the Decke basis works.  Only when the emitters is under external  force, we need to use different basis a set.

```{table} SYmmetry of a dimer
:align: center
:name: tbl:dimer-symmetry

| configuration | dipole orientation | point group | permutation |
|:--|:--|:--:|:--:|
| bare atoms (no dipole) | —                    | $D_{\infty h}$ | $S_2$ |
| dipole **along** the axis | parallel          | $C_{\infty v}$ | $S_2$ |
| dipole **across** the axis | perpendicular    | $C_{2v}$       | $S_2$ |
| dipole **tilted** | general (inclined)        | $C_{s}$        | $S_2$ |
```

We discuss three cases in details.

1. **Thermal emission from a non-interacting pair**  ($r_{12} \gg \lambda_{0}$)
2. **Thermal emission from a pair coupled by with dipole-dipole interaction:** ($r_{12} \ll \lambda_{0}$)
3. **Coherence by induced by external driving:** ($r_{12} \gg \lambda_{0}$)

## Model I: Thermal emission without the dipole-dipole interaction

When the distance between the emitters is much larger than the wave length, the dipole-dipole coupling is even smaller than $\gamma_{0}$.   Then we have a situation of the dual time scales (see {numref}`sec:openqm-qme-scaling`.)    We will eliminate the rapid oscillation of frequency $\omega_{0}$ from the standard optical master equation using a rotating frame (see {numref}`sec:openqm-qme-rotating-frame`.)

**Objectives**

1. Learn how to deal with the dual time scales using the standard optical master equation and a rotating frame.
2. Find how the coherence functions depend on the detector models.

## Model II:  Thermal emission with the dipole-dipole interaction

When the distance between the emitters is much shorter than the wavelength, the dipole-dipole interaction is no longer negligible.   Its magnitude is still negligibly small compared to the Bohr frequency $\omega_{0}$.  Hence, we don't have to worry about the broken detailed balance due to the shift of eiegen energies.  However, the magnitude of interaction is much bigger than the emission rate $\gamma_{0}$. Hence, we have a situation of triple time scales (see {numref}`sec:openqm-qme-scaling`.)  We solve the quantum master equation in a rotating frame, which requires the use of the Davies-GKSL master equation (see {numref}`sec:openqm-qme-rotating-frame`.)   

**Objectives**

1. Learn how to deal with the triple time scales using the GKSL master equation and the full interaction picture.
2. Find the effect of rapid energy exchange between the emitters in the coherence functions.
3. Find how the effect of the coupling depends on the detector models.

## Model III: Coherence induced by external driving

_still at work_

When the two emitters are driven by the same laser beam, they are "synchronized" and establish coherence in between product states $|ge\rangle$ and $|eg\rangle$.  This coherence generates interesting interference patterns, which is originally predicted by {cite:p}`Skornia2001` and experimentally observed by {cite:p}`Wolf2020`.  Their resutls are recreated numerically.  Neither product basis nor standar Dicke basis works for this system.  However, a Dicke-like basis set, which we shall call Skornia basis, exhibits super/subradiant channels, which is investigated in {cite:p}`Richter2023`.
