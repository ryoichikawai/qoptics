# Dimer Case IV: Model

We begin with the quantum master equation after eliminationg $H_{0}$ by going to the rotating frame.

$$
\frac{d}{dt} \rho = -i [H_{dd},\rho] + \sum_{j} P_{j} \mathcal{D}^{\uparrow}_{j}(\rho) + 
\sum_{ij} \Gamma_{ij} \mathcal{D}^{\downarrow}_{ij}(\rho) + \sum_{j}\frac{\gamma_{\phi}}{2} \mathcal{D}^{\phi}_{j}(\rho)
$$

where

$$
H_{dd} = \Omega_{12} (\sigma^{+}_{1} \sigma^{-}_{2} + \sigma^{+}_{2} \sigma^{-}_{1})
$$

and 

$$
\begin{aligned}
\mathcal{D}^{\uparrow}_{j}(\rho) &= \sigma^{+}_{j} \rho \sigma^{-}_{j} - \tfrac12 \{\sigma^{-}_{j}\sigma^{+}_{j},\rho\} \\
\mathcal{D}^{\downarrow}_{ij}(\rho) &= \sigma^{-}_{j} \rho \sigma^{+}_{i} - \tfrac12 \{\sigma^{+}_{i}\sigma^{-}_{j},\rho\} \\
\mathcal{D}^{\phi}_{j}(\rho) &= \sigma^{z}_{j} \rho \sigma^{z}_{j}- \tfrac12 \{\sigma^{z}_{j}\sigma^{z}_{j},\rho\} = \sigma^{z}_{j} \rho \sigma^{z}_{j}
\end{aligned}
$$

## Population dynamics

Using the product basis $|u_{1}\rangle = |ee\rangle$, $|u_{2}\rangle = |eg\rangle$, $|u_{2}\rangle = |ge\rangle$, and $|u_{4}\rangle = |gg\rangle$, and the matrix elements of the density operator as $p_{1}=\rho_{11}$, $p_{2}=\rho_{22}$, $p_{3}=\rho_{33}$, $p_{4}=\rho_{44}$, $c = \rho_{12} = c_\text{r}+ i c_\text{i}$, and $c^{*} = \rho_{21}=c_\text{r} - c_\text{i}$, the master equation can be written as a set of ODEs

$$
\begin{aligned}
\dot{p}_{1} &= -2 \gamma_{0} p_{1} + P (p_{2}+p_{3}) \\
\dot{p}_{2} &= \gamma_{0} p_{1} -(\gamma_{0}+2 P) p_{2} +P p_{4} - \Gamma_{12} c_\text{r} - 2 \Omega_{12} y \\
\dot{p}_{3} &= \gamma_{0} p_{1} -(\gamma_{0}+2 P) p_{3} +P p_{4} - \Gamma_{12} c_\text{r} + 2 \Omega_{12} y\\
\dot{p}_{4} &= \gamma_{0} (p_{2}+p_{3}) - P p_{4} + 2 \Gamma_{12} c_\text{r} \\
\dot{c}_\text{r} &= - \Gamma_{c} c_\text{r} + \Gamma_{12} \left[p_{4} - \tfrac12(p_{2}+p_{3}) \right] \\
\dot{c}_\text{i} &= - \Gamma_{c} c_\text{i} + \Omega_{12} (p_{2}-p_{3})
\end{aligned}
$$(eq:dimer-case4-qme-pop)

where $\Gamma_{c} = \gamma_{0} + P+ 2\gamma_{\phi}$.  Typical time scales are $\gamma_{\phi}/\gamma_{0} \sim 10^{6}$, $\Omega_{12}/\gamma_{0} \sim 10^{3}$, and $P/\gamma_{0} \sim 10$.  Hence these ODEs are very stiff.

## Adiabatic elimination of fast degrees

Noting that the fastest decay $\gamma{\phi}$  appears only in $\Gamma_{c}$,  we can eliminate the dynamics of $x$ and $y$ by adiabatic elimination.  Since $x$ and $y$ relax very fast, they stays near the instantaneous fixed point all time.  Following the standard adiabatic elimination method, we assume $\dot{c}_\text{r} = 0$ and $\dot{c}_\text{i} = 0$.  Hence,

$$
\begin{aligned}
c_\text{r} & = \frac{\Gamma_{12}}{\Gamma_{c}} \left[p_{4} - \tfrac12(p_{2}+p_{3})\right] \sim 10^{-6}\\
c_\text{i} & = \frac{\Omega_{12}}{\Gamma_{c}} (p_{2}-p_{3}) \sim 10^{-3}
\end{aligned}
$$(eq:dimer-case4-adiabatic-elimination)



### zeroth-order approximation

Since $c_\text{r}$ and $c_\text{i}$ are very small, we can simply ignore them , which is the zeroth order approximation. Seeting $c_\text{r}=0$ and $c_\text{i}=0$, we obtain a set of ODEs for population $p_{j}$.

$$
\begin{aligned}
\dot{p}_{1} &= -2 \gamma p_{1} + P (p_{2}+p_{3}) \\
\dot{p}_{2} &= \gamma_{0} p_{1} -(\gamma_{0}+2 P) p_{2} +P p_{4} \\
\dot{p}_{3} &= \gamma_{0} p_{1} -(\gamma_{0}+2 P) p_{3} +P p_{4} \\
\dot{p}_{4} &= \gamma_{0} (p_{2}+p_{3}) - P p_{4}
\end{aligned}
$$

which can be further simplified.  Since $y=0$, $p_{2}=p_{3}$. Taking into account normalization $p_{4} = 1 - p_{1} - p_{2} - p_{3}$, the four ODEs is reduced to the following pair

$$
\begin{aligned}
\dot{X} &= - 2 \gamma_{0} X + P Y \\
\dot{Y} &= 2 (\gamma_{0}-P) X - (\gamma_{0}+3 P) Y  + 2P
\end{aligned}
$$

where $X = p_{1}$ and $Y=p_{2}+p_{3}$.  They can be solved analytically as

$$
\begin{aligned}
X(t) = & \frac{P^2}{(P+\gamma_{0})^2}\left(1 - e^{-\gamma t}\right)^2 \\
&+ \frac{e^{-\gamma t}}{P+\gamma_{0}}  \left[(P-\gamma) e^{-\gamma t} - 2P\right] X(0)\\
&+ \frac{P}{P+\gamma_{0}} e^{-\gamma t} (e^{-\gamma t}-1) Y(0)
\end{aligned}
$$

$$
\begin{aligned}
Y(t) = & \frac{2P}{(P+\gamma_{0})^2} \left(1 - e^{-\gamma t}\right) \left(\gamma_{0} + e^{-\gamma t} P\right) \\
& + \frac{P-\gamma_{0}}{P+\gamma_{0}} e^{-\gamma t}(e^{-\gamma t}-1) X(0)\\
& + \frac{e^{-\gamma t}}{P+\gamma_{0}} \left(P-\gamma_{0} - 2P e^{-\gamma t}\right) Y(0)
\end{aligned}
$$

where $\gamma = P + \gamma_{0}$.  W can recover the original population variables as
$p_{1}=X$, $p_{2}=p_{3}=\tfrac12 Y$, $p_{4} = 1 - X- Y$.

The steady state population is found to be

$$
p^\text{ss}_{1} = \frac{P^2}{(P+\gamma_{0})^2}, \qquad  p^\text{ss}_{2} = p^\text{ss}_{3} = \frac{P \gamma_{0}}{(P+\gamma_{0})^2}, \qquad p^\text{ss}_{4} = \frac{\gamma_{0}^2}{(P+\gamma_{0})^2}
$$

In principle, we can evalutate coherence functions using these solutions but the final expression is complicated. We resort to a numerical method.

### First-order Approximation

In the first order approximation, we just plug {eq}`eq:dimer-case4-adiabatic-elimination` to the population dynamics of $p_{j}$ in Eq. {eq}`eq:dimer-case4-qme-pop`.  The resulting equation of motions are

$$
\begin{aligned}
\dot{p}_{1} &= - 2\gamma_{0} p_{1} + P (p_{2}+p_{3}) \\
\dot{p}_{2} &= \gamma_{0} p_{1} +P p_{4} - (\gamma_{0}+P) p_{2} - \frac{\Gamma_{12}^2}{\Gamma_{c}}\left(p_{4}-\frac{p_{2}+p_{3}}{2}\right) - \frac{2\Omega_{12}^2}{\Gamma_{c}} (p_{2}-p_{2}) \\
\dot{p}_{3} &= \gamma_{0} p_{1} +P p_{4} - (\gamma_{0}+P) p_{3} - \frac{\Gamma_{12}^2}{\Gamma_{c}}\left(p_{4}-\frac{p_{2}+p_{3}}{2}\right) + \frac{2\Omega_{12}^2}{\Gamma_{c}} (p_{2}-p_{2}) \\
\dot{p}_{4} &= -2P p_{4} + \gamma_{0}(p_{2}+p_{3}) + \frac{2\Gamma_{12}^2}{\Gamma_{c}} \left(p_{4}-\frac{p_{2}+p_{3}}{2}\right)
\end{aligned}
$$

which is purely equations of population dynamics and no coherence equation is involved.  Introducing $p_{\pm} = \tfrac12\left(p_{2}\pm p_{3}\right)$,  the equations are simplified further:

$$
\begin{aligned}
\dot{p}_{1} &= - 2 \gamma_{0} p_{1} + 2 P p_{+} \\
\dot{p}_{4} &= 2 \left(\frac{\Gamma_{12}^2}{\Gamma_{c}}-P\right) p_{4} + 2 \left(\gamma_{0}-\frac{\Gamma_{12}^2}{\Gamma_{c}}\right) p_{+} \\
\dot{p}_{+} &= \gamma_{0} p_{1} - \frac{\Gamma_{12}^2}{\Gamma_{c}} p_{4} + \left(\frac{\Gamma_{12}^2}{\Gamma_{c}} - \gamma_{0}- P\right) p_{+}
\end{aligned}
$$

with the normalization $p_{1}+p_{4}+2 p_{+}=1$.  The equation of motion for $p_{-}$ is decoupled from other populations which can be solved immediately as

$$
\dot{p}_{-} = -\frac{\Omega_{12}^2}{\Gamma_{c}} p_{-}\qquad \Rightarrow \qquad p_{-}(t) = p_{-}(0) e^{-2 (\Omega_{12}^2/\Gamma_{c}) t} \quad \rightarrow \quad 0
$$

Since $p_{-}$ is independent and rapidly decay to zero, we can ignore it.  Hence, we still have $p_{2}=p_{3}$ even in the first-order approximation.  Furthermore, the effect of the dipole-dipole coupling is diminished by the dephasing.
