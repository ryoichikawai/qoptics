# Dimer I: Thermal fluorescence 

We consider Kirchhoff radiation, a hypothetical situation mainly for learning purpose.  Consider a pair of two-level emitters with a transition energy in infrared region $\sim 0.1 $ eV. The pair is placed in a very hot photon gas at temperature about $1000 $ K $ \sim 0.086 $ eV.   The temperature is high enough to excite the emitters which in turn spontaneously emits infrared radiation.   The pair is well separated such that a detector can distinguish the source of radiation.  We are interested in the coherence properties, which depends on the detection methods.  WE try two different models of detection operators, one distinguishes the source and the other not.  The latter exhibits the Hanbury Brown and Twiss (HBT) effect.



## Parameter values

* $\omega_{0} = 1$
* $T = 0.86$
* $\gamma_{0} = 10^{-8}$
* $r_{12} = 2 \lambda_0$


## Detection

The two emitters are well separated and the detector can distinguish them.  We consider two cases, in one measurement, the detector monitors photons from each emitter separately and combine the two photon counts after the measurements, meaning that there is no interference between the radiations from different source.  In the second case, the photons from two sources are navigated to a detector with a narrow aperture so that their phase relation remains constant within the detector.  There is no reason that the coherency is lost in this case.
