# Trimer Case 1:  Incoherent cluster

When three atoms are well separated, the off-diagonal $\Gamma$ and the dipole-dipole interaction is negligibly small.  The situation is essentially the same as the incoherent dimer discussed in {numref}`ch:dimer-case1`.  The chance that two photons accidentally arrive at the same time increases and the no-delay second-order coherence increases from $\tfrac12$ to $\tfrac23$.
